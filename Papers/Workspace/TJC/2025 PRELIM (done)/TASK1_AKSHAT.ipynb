```python
# Task 1.1
def task1_1(artefacts_list):
    valid = []
    for i, artefact in enumerate(artefacts_list):
        if not artefact[0][0].isalpha() or not artefact[0][0].isupper() or int(''.join(artefact[0][1:])) not in range(1000, 10000):
            print(f'Invalid artefactID for {artefact[0]}')
            continue

        if int(artefact[1]) not in range(1, 101):
            print(f'Invalid rating for {artefact[0]}')
            continue

        valid.append(artefact)

    return valid

data = [("A1234", 85),("b1234", 85),("C1234", 85),("B0567", 92),("A3891", 145),("C4012", 78),("B5678", 95),("D6543", 88),("E7890", 65),("A8123", 73),("C9012", 98),("B1567", 52)]

arr = task1_1(data)
print(len(arr))
print(arr[:5])
```

```python
# Task 1.2
def quicksort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[0][1]

    less = [i for i in arr if i[1] < pivot]
    more = [i for i in arr if i[1] > pivot]
    equal = [i for i in arr if i[1] == pivot]

    return quicksort(more) + equal + quicksort(less)

def task1_2(artefacts_list):
    return quicksort(artefacts_list)

task1_2(arr)
```

```python
# Task 1.3
def bin_search(arr, rating, left, right):
    if left > right:
        return []

    mid = (left + right) // 2

    if arr[mid][1] == rating:
        results = [arr[mid]]

        i = mid - 1
        while i >= 0 and arr[i][1] == rating:
            results.append(arr[i])
            i -= 1

        i = mid + 1
        while i < len(arr) and arr[i][1] == rating:
            results.append(arr[i])
            i += 1

        return task1_2(results)

    elif rating > arr[mid][1]:
        return bin_search(arr, rating, left, mid-1)

    else:
        return bin_search(arr, rating, mid+1, right)

def task1_3(artefacts_list, target_rating):
    sorted_arr = task1_2(artefacts_list)
    return bin_search(sorted_arr, target_rating, 0, len(artefacts_list)-1)

test = task1_3(arr, 85)
print('85 Not found' if not test else test)

test = task1_3(arr, 100)
print('100 Not found' if not test else test)
```

```python
# Task 1.4
def task1_4(artefacts_list):
    if not artefacts_list or len(artefacts_list) == 0:
        return
    
    avg_rat = round(sum([i[1] for i in artefacts_list]) / len(artefacts_list), 2)
    above_avg = [i for i in artefacts_list if i[1] > avg_rat]
    bottom_3 = sorted(artefacts_list, key=lambda x: x[1])[:3]

    return (avg_rat, above_avg, bottom_3)

task1_4(arr)
```
