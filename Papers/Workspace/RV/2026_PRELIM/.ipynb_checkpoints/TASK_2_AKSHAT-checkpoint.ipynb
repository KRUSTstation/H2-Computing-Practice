```python
# Task 2.1
import csv

def task2_1(filename):
    with open(f'./resource folder/{filename}') as f:
        data = list(csv.reader(f))[1:]
        data = [[int(j) if j.isdigit() else j for j in i] for i in data]

    return data

players = task2_1("strikers.txt")
for player in players:
    print(player)
```

```python
# Task 2.2
def task2_2(strikers):
    return [i[0] for i in strikers if sum(i[2:5]) == 12]

players = task2_2(task2_1("strikers.txt"))
print(players)
```

```python
# Task 2.3
def task2_3(strikers):
    arr = []

    for player in strikers:
        goals_and_assisted = player[6] + player[7]

        if not arr:
            arr.append([player[0], goals_and_assisted])
            continue

        if goals_and_assisted > arr[0][1]:
            arr = []
        elif goals_and_assisted == arr[0][1]:
            arr.append([player[0], goals_and_assisted])

    return arr

players = task2_3(task2_1("strikers.txt"))
print(players)
```

```python
# Task 2.4
def merge(left, right):
    arr = []
    
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i][1] > right[j][1]:
            arr.append(left[i])
            i += 1
        else:
            arr.append(right[j])
            j += 1

    arr.extend(left[i:])
    arr.extend(right[j:])

    return arr

def mergesort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = mergesort(arr[mid:])
    right = mergesort(arr[:mid])

    return merge(left, right)

def task2_4(strikers):
    schools = {}

    for player in strikers:
        if player[1] not in schools.keys():
            schools[player[1]] = [player[6]]
        else:
            schools[player[1]].append(player[6])

    avg = mergesort([(k, sum(v) / len(v)) for k,v in schools.items()])

    print(f"{'School':<20} {'Average Goals':<20}")
    print(f'{'-' * 40}')
    
    for item in avg:
        print(f"{item[0]:<20} {item[1]:<20.2f}")
    
task2_4(task2_1("strikers.txt"))
```
