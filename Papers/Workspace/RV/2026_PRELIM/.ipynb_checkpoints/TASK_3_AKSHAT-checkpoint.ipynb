```python
# Task 3.1
def task3_1():
    letters = []
    
    for i in range(5):
        valid = False

        while not valid:
            valid = True
            
            new_l = input(f'Enter substring no. {i+1}')
            if len(new_l) != 5:
                print("You must type 5 letters each time.")
                valid = False
                
            if not new_l.isalpha():
                print("You must only input letters.")
                valid = False

            if len(set(new_l)) != len(new_l):
                print("You must only input unique letters")
                valid = False

            if len([True for i in new_l if i in [j for j in i for i in letters]]) != 0:
                letters = [i for i in new_l if i in [j for j in m for m in letters]]
                print(f"You have keyed in {''.join(letters)} before.")
                valid = False

        print('Success!')
        letters.append([i for i in new_l])

    return letters

task3_1()
```
