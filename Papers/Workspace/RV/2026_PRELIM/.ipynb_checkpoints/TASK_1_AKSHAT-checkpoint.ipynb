```python
# Task 1.1
class Tile:
    def __init__(self, tile_id):
        self.tile_id = tile_id
        self.next_tile = None

    def get_tile_id(self):
        return self.tile_id
    def set_tile_id(self, data):
        self.tile_id = data

    def get_next_tile(self):
        return self.next_tile
    def set_next_tile(self, data):
        self.next_tile = data

    def __str__(self):
        return f'N{self.get_tile_id()}'

class Ladder(Tile):
    def __init__(self, tile_id):
        super().__init__(tile_id)
        self.ladder_tile = None

    def get_ladder_tile(self):
        return self.ladder_tile
    def set_ladder_tile(self, data):
        self.ladder_tile = data

    def __str__(self):
        if not self.get_ladder_tile():
            return f'L{self.get_tile_id()}'

        return f'L{self.get_tile_id()} goes to {self.get_ladder_tile().get_tile_id()}'

class Snake(Tile):
    def __init__(self, tile_id):
        super().__init__(tile_id)
        self.snake_tile = None

    def get_snake_tile(self):
        return self.snake_tile
    def set_snake_tile(self, data):
        self.snake_tile = data

    def __str__(self):
        if not self.get_snake_tile():
            return f'S{self.get_tile_id()}'

        return f'S{self.get_tile_id()} goes to {self.get_snake_tile().get_tile_id()}'
```

```python
# Task 1.2
def create_path(path_str):
    first = None
    previous = None
    for i, char in enumerate(path_str):
        if i < 10:
            i = '0' + str(i)
        else:
            i = str(i)
        
        if char == 'N':
            node = Tile(i)
        elif char == 'L':
            node = Ladder(i)
        else:
            node = Snake(i)
        
        if previous:
            previous.next_tile = node
        else:
            first = node

        previous = node

    return first

path = create_path("NLNS")
print(path)
print(path.next_tile)
print(path.next_tile.next_tile.next_tile)
```

```python
# Task 1.3
def display_tiles(path):
    current = path
    while current:
        print(current)
        current = current.next_tile

def test1_3():
    path = create_path("NLNNNNSNNNNLNNN")
    display_tiles(path)

test1_3()
```

```python
# Task 1.4
def find_next_ladder(tile):
    current = tile
    while current:
        if isinstance(current, Ladder):
            return current
        current = current.next_tile

    return

def find_next_snake(tile):
    current = tile
    while current:
        if isinstance(current, Snake):
            return current
        current = current.next_tile

    return

def test1_4():
    path = create_path("NLNNNNSSNNNLNNN")
    curr = path

    curr = find_next_ladder(curr)
    print(curr)

    curr = find_next_snake(curr)
    print(curr)

    curr = find_next_ladder(curr)
    print(curr)

test1_4()
```

```python
# Task 1.5
def move_forward(tile, n):
    count = 0
    current = tile
    
    while current.next_tile and count < n:
        current = current.next_tile
        count += 1

    return current

def test1_5():
    path = create_path("NLNNNNSNNNNLNNN")
    curr = path

    curr = move_forward(curr, 2)
    print(curr)

    curr = move_forward(curr, 6)
    print(curr)

    curr = move_forward(curr, 20)
    print(curr)

test1_5()
```

```python
# Task 1.6
def activate_ladder_tiles(path, ladder_str):
    destinations = ladder_str.split(',')[1:]
    destinations = [int(i) for i in destinations]

    count = 0
    ladder = find_next_ladder(path)
    current = path

    while destinations:
        if count == destinations[0]:
            ladder.set_ladder_tile(current)

            destinations.pop(0)
            ladder = find_next_ladder(current.next_tile)
        
        current = current.next_tile
        count += 1

def test1_6():
    path = create_path("NNLNNSNNNLNNSNN")
    activate_ladder_tiles(path, "L,3,13")
    display_tiles(path)

test1_6()
```

```python
# Task 1.7
def activate_snake(path, snake_str):
    destinations = snake_str.split(',')[1:]
    destinations = [int(i) for i in destinations]

    count = 0
    snake = find_next_snake(path)
    current = path

    while destinations:
        if count == destinations[0]:
            snake.set_snake_tile(current)

            destinations.pop(0)
            snake = find_next_snake(snake.next_tile)
        
        current = current.next_tile
        count += 1

def test1_7():
    path = create_path("NNLNNSNLNLNNSNN")
    activate_snake(path, "S,3,6")
    display_tiles(path)

test1_7()
```
