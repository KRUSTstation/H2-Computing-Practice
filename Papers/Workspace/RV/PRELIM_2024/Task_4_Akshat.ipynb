```python
# Task 4.1
class Contestant:
    def __init__(self, name, time):
        self.name = name
        self.time = time

    def get_time(self):
        return self.time

    def set_time(self, new_time):
        self.time = new_time

    def __str__(self):
        name = self.name.replace(' ', '')
        return f"C-{''.join([name[i].upper() if i < len(name) else '_' for i in range(8)])}({self.get_time()}s)"

class Beginner(Contestant):
    def __init__(self, name, time, practice_hours):
        super().__init__(name, time)
        self.practice_hours = practice_hours

    def __str__(self):
        name = self.name.replace(' ', '')
        return f"B-{''.join([name[i].upper() if i < len(name) else '_' for i in range(8)])}({self.get_time()}s)"

class Professional(Contestant):
    def __init__(self, name, time, competitions_won):
        super().__init__(name, time)
        self.competitions_won = competitions_won

    def __str__(self):
        name = self.name.replace(' ', '')
        return f"P-{''.join([name[i].upper() if i < len(name) else '_' for i in range(8)])}({self.get_time()}s)"

print(str(Contestant("ChanChan Chan",35.01))) 
print(str(Beginner("Li uLi uLi u",55.54, 300))) 
print(str(Professional("OhOhOh",6.34, 60)))
```

```python
# Task 4.2
def readfile(filename):
    to_return = []

    with open(filename) as f:
        data = [i.strip().split(',') for i in f.readlines()]
        for item in data:
            if item[0] == 'Beginner':
                to_return.append(Beginner(*item[1:]))

            elif item[0] == 'Professional':
                to_return.append(Professional(*item[1:]))

    return to_return
```

```python
# Task 4.3
class Node:
    def __init__(self, player):
        if not isinstance(player, Beginner) and not isinstance(player, Professional):
            return

        self.player = player
        self.left = None
        self.right = None

class BST:
    def __init__(self):
        self.root = None

    def insert(self, player):
        if not isinstance(player, Beginner) and not isinstance(player, Professional):
            return

        self.root = self.insert_helper(self.root, player)

    def insert_helper(self, root, key):
        if root is None:
            return Node(key)

        if key.get_time() < root.player.get_time():
            root.left = self.insert_helper(root.left, key)

        elif key.get_time() > root.player.get_time():
            root.right = self.insert_helper(root.right, key)

        return root

    def inorder(self):
        result = []
        self.inorder_helper(self.root, result)
        return result

    def inorder_helper(self, root, result):
        if root:
            self.inorder_helper(root.left, result)
            result.append(root.player)
            self.inorder_helper(root.right, result)

    def count_professional(self):
        def helper(node):
            if node is None:
                return 0

            count = 1 if isinstance(node.player, Professional) else 0

            return (
                count
                + helper(node.left)
                + helper(node.right)
            )

        return helper(self.root)

    def find_best_beginner(self):
        def helper(node):
            if node is None:
                return None

            result = helper(node.right)
            if result is not None:
                return result

            if isinstance(node.player, Beginner):
                return node.player

            return helper(node.left)

        return helper(self.root)

bst = BST() 
players = readfile("rubik.txt") 
for player in players: 
    bst.insert(player) 
print(bst.inorder())
print(bst.count_professional()) 
print(bst.find_best_beginner())
```
