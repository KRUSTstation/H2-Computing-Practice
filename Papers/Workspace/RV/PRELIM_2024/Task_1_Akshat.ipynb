```python
# Task 1.1
import sqlite3

conn = sqlite3.connect('cu_booking.db')
cursor = conn.cursor()

with conn:
    cursor.executescript('''
    DROP TABLE student;
    DROP TABLE cubicle;
    DROP TABLE booking;
''')

    cursor.executescript('''
    CREATE TABLE IF NOT EXISTS student(
        id TEXT PRIMARY KEY,
        name TEXT NOT NULL,
        contact INT NOT NULL
    );

    CREATE TABLE IF NOT EXISTS cubicle(
        cubicle_no TEXT PRIMARY KEY,
        maintenance INT NOT NULL
    );

    CREATE TABLE IF NOT EXISTS booking(
        date TEXT NOT NULL,
        cubicle_no TEXT NOT NULL,
        student_id TEXT NOT NULL,

        PRIMARY KEY (cubicle_no, date),

        FOREIGN KEY (cubicle_no) REFERENCES cubicle(cubicle_no),
        FOREIGN KEY (student_id) REFERENCES student(id)
    );
    ''')

    conn.commit()
```

```python
# Task 1.2
import csv

conn = sqlite3.connect('cu_booking.db')
cursor = conn.cursor()

with open('./student.csv') as f:
    data = [i for i in csv.reader(f)][1:]
    for i in data:
        cursor.execute('INSERT INTO student(id, name, contact) VALUES (?, ?, ?)', i)

with open('./cubicle.csv') as f:
    data = [i for i in csv.reader(f)][1:]
    for i in data:
        cursor.execute('INSERT INTO cubicle(cubicle_no, maintenance) VALUES (?, ?)', i)

with open('./booking.csv') as f:
    data = [i for i in csv.reader(f)][1:]
    for i in data:
        cursor.execute('INSERT INTO booking(date, cubicle_no, student_id) VALUES (?, ?, ?)', i)

conn.commit()
conn.close()
```

```python
# Task 1.5
from flask import Flask, render_template, url_for, redirect

app = Flask(__name__, template_folder='./Task_1_5_Akshat')

@app.route('/')
def base():
    return redirect(url_for('availability', date='17112024'))

@app.route('/availability/<date>', methods=['GET'])
def availability(date):
    conn = sqlite3.connect('cu_booking.db')
    cursor = conn.cursor()

    with conn:
        cursor.execute('SELECT cubicle_no FROM cubicle WHERE maintenance = 0')
        cubicles = cursor.fetchall()
        valid_cubicles = []

        for i in cubicles:
            cursor.execute('SELECT * FROM booking WHERE cubicle_no = ? AND date = ?', (i[0], date))
            if not cursor.fetchone():
                valid_cubicles.append(i[0])

    return render_template('base.html', cubicles = cubicles, valid = valid_cubicles, date = date)

app.run()
```
