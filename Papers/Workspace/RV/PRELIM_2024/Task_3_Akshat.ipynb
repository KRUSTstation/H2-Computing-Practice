```python
# Task 3.1
import random

def create_puzzle(n):
    arr = [[None for _ in range(n)] for _ in range(n)]
    numbers = [i for i in range(n**2)]
    random.shuffle(numbers)
    r = None

    count = 0
    for i in range(len(arr)):
        for j in range(len(arr[i])):
            arr[i][j] = numbers[count]
            if arr[i][j] == 0:
                r = (i, j)
            count+=1

    return arr, r[0], r[1]

for i in range(3, 6):
    puzzle, r, c = create_puzzle(i) 
    print("Coordinate of space: ({},{})".format(r,c)) 
    print("Puzzle: " + str(puzzle)) 
    print()
```

```python
# Task 3.2
def display_puzzle(puzzle):
    print(f"{'':>5}", end='')
    for i in range(len(puzzle)):
        print(f"{i:>5}", end='')
    print()

    for i in range(len(puzzle)):
        print(f"{i:>5}", end='')
        for j in range(len(puzzle[i])):
            if puzzle[i][j] == 0:
                print(f"{'#':>5}", end='')
                continue
            print(f"{puzzle[i][j]:>5}", end='')
        print()

puzzle = [[14, 10, 9, 3], [2, 4, 5, 12], [13, 0, 15, 8], [7, 1, 6, 11]] 
display_puzzle(puzzle)            
```

```python
# Task 3.3
def find_available_move(puzzle, r, c):
    available_moves = ['U', 'R', 'D', 'L']
    if c == 0:
        available_moves.remove('L')
    elif c == len(puzzle) - 1:
        available_moves.remove('R')

    if r == 0:
        available_moves.remove('U')
    elif r == len(puzzle) - 1:
        available_moves.remove('D')

    return available_moves

puzzle = [[14, 10, 9, 3], [2, 4, 5, 12], [13, 0, 15, 8], [7, 1, 6, 11]] 
find_available_move(puzzle, 2, 1)
```

```python
# Task 3.4
def make_move(move, puzzle, r, c):
    if move =='R':
        puzzle[r][c], puzzle[r][c+1] = puzzle[r][c+1], puzzle[r][c]

    elif move == 'L':
        puzzle[r][c], puzzle[r][c-1] = puzzle[r][c-1], puzzle[r][c]

    elif move == 'U':
        puzzle[r][c], puzzle[r-1][c] = puzzle[r-1][c], puzzle[r][c]

    elif move == 'D':
        puzzle[r][c], puzzle[r+1][c] = puzzle[r+1][c], puzzle[r][c]

puzzle = [[14, 10, 9, 3], [2, 4, 5, 12], [13, 0, 15, 8], 
[7, 1, 6, 11]] 
make_move('D', puzzle, 2, 1) 
print(puzzle)
```

```python
# Task 3.5
def win(puzzle):
    prev = None

    for i in puzzle:
        for j in i:
            if not prev:
                prev = j
                continue

            if j-prev == 1 and puzzle[len(puzzle)-1][len(puzzle)-1] == 0:
                continue

            return False
            
    return True
```

```python
# Task 3.6
import time

def game_menu():
    def display_menu():
        display_puzzle(puzzle)
        print('''
            1) New game
            2) Make one move
            3) Make many moves
            4) Quit
        ''')

    def find_root():
        for i in range(len(puzzle)):
            for j in range(len(puzzle[i])):
                if puzzle[i][j] == 0:
                    return i, j

    puzzle, r, c = [[2,3,5], [1,0,4], [7,8,6]], 1, 1

    start_time = time.perf_counter()

    while not win(puzzle):
        display_menu()
        option = input()

        if option == '1':
            data = None
            while data not in ['3', '4', '5']:
                data = input('Choose a size between 3 to 5')

            puzzle, r, c = create_puzzle(int(data))
            
            start_time = time.perf_counter()

        elif option == '2':
            print('Valid moves: ', end='')
            valid_moves = find_available_move(puzzle, r, c)
            print(valid_moves)

            selection = None
            while selection not in valid_moves:
                selection = input('Which move?')

            make_move(selection, puzzle, r, c)
            r, c = find_root()

        elif option == '3':
            og_puzzle = []
            for i in range(len(puzzle)):
                og_puzzle[i] = []
                for j in range(len(puzzle[i])):
                    og_puzzle[i].append(puzzle[i][j])

            moves = input('Type moves')
            for i in moves:
                valid_moves = find_available_move(puzzle, r, c)
                if i not in valid_moves:
                    puzzle = og_puzzle
                    print('invalid!')
                    continue

                make_move(i, puzzle, r, c)
                r, c = find_root()

        elif option == '4':
            end_time = time.perf_counter()
            t = end_time - start_time

            print(f'You took {t}s to give up')
            return

    end_time = time.perf_counter()
    t = end_time - start_time
    print(f'You won in {t}s!!')

game_menu()
```
