```python
# Task 2.1
def task_2_1(filename):
    L = []

    with open(f'./{filename}') as f:
        for i in f.read().split(', '):
            L.append(chr(int(i)))

    return L

task_2_1("inputA.txt")
```

```python
# Task 2.2
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j][1] < key[1]:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr

def task_2_2(char_list):
    freq = {}
    for i in char_list:
        if not freq.get(i):
            freq[i] = 1
        else:
            freq[i] += 1

    sorted_freq = insertion_sort(list(freq.items()))

    print(f'''
    {'Char':>5} {'Freq':>5}
''', end='')
    for i in sorted_freq:
        print(f"{'':>5}'{i[0]}' {i[1]:5}")

task_2_2(task_2_1("inputA.txt"))
```

```python
# Task 2.3
def task_2_3(char_list):
    n = int(len(char_list) ** 0.5)
    arr = [[None] * n for _ in range(n)]

    top = 0
    bottom = n - 1
    left = 0
    right = n - 1

    index = [0, 0]
    direction = "down"

    i = 0
    while i < len(char_list):

        arr[index[0]][index[1]] = char_list[i]
        i += 1

        if direction == "down":
            if index[0] < bottom:
                index[0] += 1
            else:
                direction = "right"
                left += 1
                index[1] += 1

        elif direction == "right":
            if index[1] < right:
                index[1] += 1
            else:
                direction = "up"
                bottom -= 1
                index[0] -= 1

        elif direction == "up":
            if index[0] > top:
                index[0] -= 1
            else:
                direction = "left"
                right -= 1
                index[1] -= 1

        elif direction == "left":
            if index[1] > left:
                index[1] -= 1
            else:
                direction = "down"
                top += 1
                index[0] += 1

    return arr


result = task_2_3('#abcdefghijklmn@')
print(result)
result = task_2_3(task_2_1("example.txt")) 
for row in result: 
    for char in row: 
        print(char, end = "") 
    print()
```

```python
# Task 2.4
def task_2_4(input_file, outputfile):
    result = task_2_3(task_2_1(input_file))

    with open(outputfile, 'w') as f:
        for row in result: 
            f.writelines(''.join(row) + '\n')

task_2_4("example.txt", "output.txt") 
task_2_4("inputA.txt", "outputA.txt") 
task_2_4("inputB.txt", "outputB.txt")
```
