```python
# Task 2.1
def task_2_1(filename):
    L = []

    with open(f'./{filename}') as f:
        for i in f.read().split(', '):
            L.append(chr(int(i)))

    return L

task_2_1("inputA.txt")
```

```python
# Task 2.2
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j][1] < key[1]:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr

def task_2_2(char_list):
    freq = {}
    for i in char_list:
        if not freq.get(i):
            freq[i] = 1
        else:
            freq[i] += 1

    sorted_freq = insertion_sort(list(freq.items()))

    print(f'''
    {'Char':>5} {'Freq':>5}
''', end='')
    for i in sorted_freq:
        print(f"{'':>5}'{i[0]}' {i[1]:5}")

task_2_2(task_2_1("inputA.txt"))
```

```python
# Task 2.3
def task_2_3(char_list):
    n = int(len(char_list) ** 0.5)
    arr = [[None] * n for _ in range(n)]

    index = [0, 0]
    boundary = n - 1
    direction = 'down'

    i = 0
    while i < len(char_list):
        char = char_list[i]

        if direction == 'down':
            if index[0] >= boundary:
                direction = 'right'
                continue

            index[0] += 1
            arr[index[0]][index[1]] = char

        elif direction == 'right':
            if index[1] >= boundary:
                direction = 'up'
                boundary -= 1
                continue

            index[1] += 1
            arr[index[0]][index[1]] = char

        elif direction == 'up':
            if index[0] <= boundary:
                direction = 'left'
                continue

            index[0] -= 1
            arr[index[0]][index[1]] = char

        elif direction == 'left':
            if index[1] <= boundary:
                direction = 'down'
                boundary -= 1
                continue

            index[1] -= 1
            arr[index[0]][index[1]] = char

        i += 1
        
    # count = 0
    # for i in range(len(arr)):
    #     for j in range(len(arr[i])):
    #         arr[i][j] = char_list[count]
    #         count+=1

    return arr

result = task_2_3('#abcdefghijklmn@')
print(result)
result = task_2_3(task_2_1("example.txt")) 
for row in result: 
    for char in row: 
        print(char, end = "") 
    print()
```
