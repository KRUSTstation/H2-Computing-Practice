```python
# Task 4.1
def task4_1(main_text, keyword):
    if len(main_text) <= 0 or len(keyword) <= 0: return

    final = ''
    alpha_stuff = ''

    for i, char in enumerate(main_text):
        if not char.isalpha() or char == ' ':
            final += char
            continue
        
        if len(keyword) <= i:
            #print(final[alpha_c])
            shift = ord(alpha_stuff[-1]) - ord('a')
        else:
            shift = ord(keyword[i]) - ord('a')

        if char.isupper():
            final += chr((ord(char.lower()) - ord('a') + shift) % 26 + ord('a')).upper()
            alpha_stuff += chr((ord(char.lower()) - ord('a') + shift) % 26 + ord('a')).upper()
        else:
            final += chr((ord(char.lower()) - ord('a') + shift) % 26 + ord('a'))
            alpha_stuff += chr((ord(char.lower()) - ord('a') + shift) % 26 + ord('a'))

    return final

print(task4_1('Hello everybody!', 'key'))
```
