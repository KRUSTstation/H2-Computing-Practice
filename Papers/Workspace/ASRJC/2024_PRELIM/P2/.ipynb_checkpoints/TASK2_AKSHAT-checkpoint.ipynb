```python
# Task 2.1
class BSTNode:
    def __init__(self, ds, score, data):
        self.right = None
        self.left = None
        self.ds = ds
        self.score = score
        if self.ds == 'S':
            self.data = Stack()
            self.data.push(data)
        else:
            self.data = Queue()
            self.data.enqueue(data)

class BST:
    def __init__(self):
        self.root = None

    def insert(self, ds, score, data):
        self.root = self.insert_helper(self.root, ds, score, data)

    def insert_helper(self, root, ds, score, data):
        if root is None:
            return BSTNode(ds, score, data)

        if score < root.score:
            root.left = self.insert_helper(root.left, ds, score, data)

        elif score > root.score:
            root.right = self.insert_helper(root.right, ds, score, data)

        else:
            if root.ds == 'S':
                root.data.push(data)
            else:
                root.data.enqueue(data)

        return root

    def search(self, score):
        return self.search_helper(self.root, score)

    def search_helper(self, root, score):
        if root is None or root.score == score:
            return root

        if score < root.score:
            return self.search_helper(root.left, score)
        else:
            return self.search_helper(root.right, score)

    def inorder(self):
        result = []
        self.inorder_helper(self.root, result)
        return result

    def inorder_helper(self, root, result):
        if root:
            self.inorder_helper(root.left, result)
            result.append(root.score)
            self.inorder_helper(root.right, result)

class LinkedListNode:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.first = None

    def insert(self, data):
        node = LinkedListNode(data)
        if not self.first:
            self.first = node
            return

        current = self.first
        while current.next:
            current = current.next

        current.next = node

class Stack:
    def __init__(self):
        self.LL = LinkedList()
        self.top = -1
        self.size = 0

    def is_empty(self):
        return self.size == 0

    def is_full(self):
        return self.size == 10
    
    def push(self, data):
        if self.is_full(): return
        
        self.LL.insert(data)
        self.size += 1

    def pop(self):
        if self.is_empty(): return

        prev = None
        current = self.LL.first
        while current.next:
            prev = current
            current = current.next

        prev.next = None
        self.size -= 1

    def display(self):
        current = self.LL.first
        while current:
            print(current.data, end="->")
            current = current.next

        print("None")

class Queue:
    def __init__(self):
        self.arr = [None] * 10
        self.front = 0
        self.rear = -1
        self.size = 0

    def is_empty(self):
        return self.size == 0

    def is_full(self):
        return self.size == 10

    def enqueue(self, data):
        if self.is_full(): return

        self.rear = (self.rear+1) % 10
        self.arr[self.rear] = data
        self.size += 1

    def dequeue(self):
        if self.is_empty(): return

        front = self.arr[self.front]
        self.front = (self.front + 1) % 10
        self.size -= 1
        return front

    def display(self):
        print([i for i in self.arr if i is not None])
```

```python
# Task 2.2
import csv

bst = BST()

with open('./RESOURCES/scores.txt') as f:
    data = [i for i in csv.reader(f)]
    for i in data:
        bst.insert(i[0], int(i[1]), i[2])

    print(bst.inorder())
```

```python
# Task 2.3
items = [bst.search(75), bst.search(53)]

for i in items:
    if i is not None:
        print(i.data.display())
    else:
        print("Not found")
```

```python
def maximum(bst):
    current = bst.root
    prev = current
    
    while current is not None:
        prev = current
        current = current.right

    prev.data.display()

maximum(bst)
```
