```python
# Task 1.1
with open('./RESOURCES/votes.txt') as f:
    data = [i.strip().split(' ') for i in f.readlines()]
    votes = {}
    for _, person in data:
        if person not in votes.keys():
            votes[person] = 1
        else:
            votes[person] += 1

    results = [[person, vote] for person, vote in votes.items()]

print(results)
```

```python
# Task 1.2
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i-1

        while j>=0 and arr[j][1] < key[1]:
            arr[j+1] = arr[j]
            j-=1

        arr[j+1] = key

    return arr

print(insertion_sort(results))

# part 2
sorted_arr = insertion_sort(results)
highest = [i for i in sorted_arr if i[1] == sorted_arr[0][1]]
print(highest)
```

```python
# Task 1.3
def binary_search(arr, t):
    left = 0
    right = len(arr) - 1

    while left <= right:
        mid = (left + right) // 2

        #print(arr[mid])
        if arr[mid][1] == t:
            return mid

        elif arr[mid][1] > t:
            left = mid + 1

        else:
            right = mid - 1

    return -1
            
def task1_3(t):
    curr_arr = sorted_arr
    found_arr = []
    found = True

    while found:
        index = binary_search(curr_arr, t)
        if index >= 0:
            found_arr.append(curr_arr[index])
            curr_arr.pop(index)
        else:
            found = False

    return found_arr
        
print(task1_3(16))
print(task1_3(2))
```
