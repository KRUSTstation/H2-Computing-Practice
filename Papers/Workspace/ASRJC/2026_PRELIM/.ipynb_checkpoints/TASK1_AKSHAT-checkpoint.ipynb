```python
# Task 1.1
def is_valid(code):
    if isinstance(code, str) and len(code) == 4:
        for char in code:
            if char not in ["R", "Y", "G", "B"]:
                return False
        return True
    return False

print(is_valid("RYGY"))
print(is_valid("RGB"))
print(is_valid("YYYY"))
print(is_valid("BRGE"))
print(is_valid(["R", "G", "B", "Y"]))
print(is_valid(6767))
```

```python
# Task 1.2
def score_guess(secret, guess):
    if not is_valid(secret) or not is_valid(guess):
        return

    hits = 0
    p_hits = 0 
    
    for i, char in enumerate(secret):
        if guess[i] == char:
            hits += 1
        elif char in guess:
            p_hits += 1

    return (hits, p_hits)

print(score_guess("RGBY", "GGRR"))
print(score_guess("RGGB", "RGGB"))
print(score_guess("RGGB", "YYYY"))
```

```python
# Task 1.3
import csv

def load_guesses(filename, secret):
    with open(filename) as f:
        data = [i for i in csv.reader(f)][1:]

    guesses = []
    
    for guess in data:
        item = score_guess(secret, ''.join(guess))
        value = item[0] * 10 + item[1]

        guesses.append(tuple(guess + [value]))

    return guesses

values = load_guesses("MASTERMIND_GUESSES.txt", "RGGB")
print(values)

def insertion_sort(guesses):
    for i in range(1, len(guesses)):
        key = guesses[i][4]
        j = i - 1

        while j >= 0 and guesses[j][4] > key:
            guesses[j], guesses[j+1] = guesses[j+1], guesses[j]
            j -= 1

    return guesses

print()
print(insertion_sort(values))
```

```python
# Task 1.4
def binary_search(data, target, low, high):
    if low > high:
        return

    mid = (low + high) // 2

    if target < data[mid][4]:
        return binary_search(data, target, low, mid - 1)
    elif target > data[mid][4]:
        return binary_search(data, target, mid + 1, high)

    return data[mid]

values = insertion_sort(load_guesses("MASTERMIND_GUESSES.txt", "RGGB"))
print(binary_search(values, 21, 0, len(values)))
print(binary_search(values, 40, 0, len(values)))
print(binary_search(values, 15, 0, len(values)))
```
