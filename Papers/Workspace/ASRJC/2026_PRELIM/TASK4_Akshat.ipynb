```python
# Task 4.1
import csv
import sqlite3

conn = sqlite3.connect('SHENZHENGO.db')
cursor = conn.cursor()

with conn:
    cursor.executescript('''
        DROP TABLE IF EXISTS Room;
        DROP TABLE IF EXISTS Student;
    
        CREATE TABLE IF NOT EXISTS Room(
            Room_ID INTEGER PRIMARY KEY,
            Room_Type TEXT NOT NULL
        );
    
        CREATE TABLE IF NOT EXISTS Student(
            Student_ID INTEGER PRIMARY KEY,
            Student_Name TEXT NOT NULL,
            Student_Class TEXT NOT NULL,
            Student_Gender TEXT NOT NULL,
            Student_Leader INT NOT NULL,
            Room_ID INT NOT NULL,

            FOREIGN KEY (Room_ID) REFERENCES Room(Room_ID)
        );
    ''')

    with open('ROOMS.txt') as f:
        data = [i for i in csv.reader(f)][1:]
        
        for item in data:
            cursor.execute('INSERT INTO Room VALUES (?, ?)', item)

    with open('STUDENTS.txt') as f:
        data = [[1 if j == 'TRUE' else 0 if j == 'FALSE' else j for j in i] for i in csv.reader(f)][1:]
        
        for item in data:
            print(item)
            cursor.execute('INSERT INTO Student VALUES (?, ?, ?, ?, ?, ?)', item)
    
    conn.commit()
```
