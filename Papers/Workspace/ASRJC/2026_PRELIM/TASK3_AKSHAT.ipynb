```python
# Task 3.1
class Node:
    def __init__(self, data):
        self.Data = data
        self.left_ptr = None
        self.right_ptr = None

    def average(self):
        return sum(self.Data[1]) / len(self.Data[1])

    def display_info(self):
        print(f"Class group: {self.Data[0]}")
        print(f"Average GP score: {self.average()}")
```

```python
# Task 3.2 - 3.5
class GPTree:
    def __init__(self):
        self.root = None

    def insert(self, data):
        self.root = self._insert(self.root, Node(data)) 

    def _insert(self, root, data):
        if not root:
            return data

        if data.average() < root.average():
            root.left_ptr = self._insert(root.left_ptr, data)

        elif data.average() > root.average():
            root.right_ptr = self._insert(root.right_ptr, data)

        return root

    def inorder(self):
        result = []
        self._inorder(result, self.root)
        return result

    def _inorder(self, result, root):
        if root:
            self._inorder(result, root.left_ptr)
            result.append(root)
            self._inorder(result, root.right_ptr)

    def find_min(self):
        if not self.root: return
        
        return self._find_min(self.root)

    def _find_min(self, root):
        if not root.left_ptr:
            return root

        return self._find_min(root.left_ptr)

    def find_max(self):
        if not self.root: return
        
        return self._find_max(self.root)

    def _find_max(self, root):
        if not root.right_ptr:
            return root

        return self._find_max(root.right_ptr)

    def depth(self):
        return len(self.inorder())

    def tree_avg(self):
        result = self.inorder()
        scores = [i.average() for i in result]
        return sum(scores) / len(scores)

    def classes_below_thresehold(self, thresehold):
        result = []
        self._classes_below_thresehold(result, thresehold, self.root)
        return result

    def _classes_below_thresehold(self, result, thresehold, root):
        if root:
            if root.average() < thresehold:
                result.append((root.Data[0], root.average()))
            self._classes_below_thresehold(result, thresehold, root.left_ptr)
            self._classes_below_thresehold(result, thresehold, root.right_ptr)

# Task 3.3
with open("GP_SCORES.txt") as f:
    data = [i.strip().split(' : ') for i in f.readlines()]
    data = [(i[0], [int(j) for j in i[1].split(', ')]) for i in data]

tree = GPTree()

for item in data:
    tree.insert(item)

result = tree.inorder()
for item in result:
    print(f"{item.Data[0]} scored an average of {item.average()}")

# Task 3.4
mi = tree.find_min()
print(mi.Data[0], mi.average())

ma = tree.find_max()
print(ma.Data[0], ma.average())

print(tree.depth())

print(tree.tree_avg())

# Task 3.5
result = tree.classes_below_thresehold(25)
print(f"Thresehold of {25}")
for i in result:
    print(i)
print(f"Total: {len(result)}")
print()

result = tree.classes_below_thresehold(27)
print(f"Thresehold of {27}")
for i in result:
    print(i)
print(f"Total: {len(result)}")
```
