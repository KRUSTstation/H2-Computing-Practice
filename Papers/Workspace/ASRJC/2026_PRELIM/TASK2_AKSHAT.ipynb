```python
# Task 2.2
from pymongo import MongoClient
from datetime import datetime

client = MongoClient('localhost', 27017)
db = client['paws_haven']
coll = db['animal']

def display(items):
    print(f"Number: {len(items)}")
    for i in items:
        print(f"Animal_id: {i['animal_id']}")
        print(f"Name: {i['name']}")
        print(f"Species: {i['species']}")
        print()

# Comb 1
items = list(coll.find({
    'species': 'Cat',
    'vaccinated': True,
    'status': 'Left'
}))

desired = []

for cat in items:
    date_left = datetime.strptime(cat['date_left'], '%Y-%m-%d')
    if datetime.strptime('2025-03-01', '%Y-%m-%d') <= date_left <= datetime.strptime('2026-01-31', '%Y-%m-%d'):
        desired.append(cat)

print("Combination 1")
display(desired)

# Comb 2
items = list(coll.find({
    'species': {"$in": ["Dog", "Rabbit"]},
    'vaccinated': False,
    'status': 'In Shelter',
    'adoption_fee': {"$gt" : 45}
}))

print("Combination 2")
display(items)
```

```python
# Task 2.3
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

    def get_data(self):
        return self.data
    def set_data(self, data):
        self.data = data

class Queue:
    def __init__(self):
        self.front = None
        self.back = None
        self.count = 0

    def size(self):
        return self.count

    def enqueue(self, data):
        if self.size() == 0:
            self.front = Node(data)
            self.back = self.front
            
        else:
            self.back.next = Node(data)
            self.back = self.back.next

        self.count += 1

    def dequeue(self):
        if self.size() == 0:
            return

        if self.size() == 1:
            to_return = self.front
            self.front = self.back = None
            
        else:
            to_return = self.front
            self.front = self.front.next
        
        self.count -= 1
        return to_return.get_data()
```

```python
# Task 2.4
class Animal:
    def __init__(self, animal_id, species, gender, arrival_date):
        self.animal_id = animal_id
        self.species = species
        self.gender = gender
        self.arrival_date = arrival_date

    def display(self):
        print(f"Animal ID: {self.animal_id}")
        print(f"Species: {self.species}")
        print(f"Gender: {self.gender}")
        print(f"Arrival Date: {self.arrival_date}")

q = Queue()

q.enqueue(Animal('PH-2001', 'Dog', 'Male', '2026-08-20'))
q.enqueue(Animal('PH-2006', 'Cat', 'Female', '2026-08-20'))
q.enqueue(Animal('PH-2007', 'Rabbit', 'Female', '2026-08-24'))
q.enqueue(Animal('PH-2009', 'Dog', 'Male', '2026-08-25'))

print(q.size())

while q.size() > 0:
    a = q.dequeue()

    print()
    a.display()

print()
a = q.dequeue()
print(a)
```
