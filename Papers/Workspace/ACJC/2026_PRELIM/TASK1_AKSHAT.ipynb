```python
# Task 1.1
def factor_sum(x):
    divisor = 1

    factor_sum = 0
    while divisor <= x/2:
        if x % divisor == 0:
            factor_sum += divisor
        divisor += 1

    return factor_sum

factor_sum(220)
```

```python
# Task 1.2
k = int(input('Give k: '))

i = j = 1
while j <= k:
    if factor_sum(i) == i:
        j += 1
        print(i)

    i += 1
```
