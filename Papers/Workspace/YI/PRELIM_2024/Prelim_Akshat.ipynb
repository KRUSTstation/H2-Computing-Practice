```python
# Task 1.1
def hash(address):
    temp = 9619
    for char in address:
        temp *= 37
        temp += ord(char)

    return temp % 250

hash('ABC')
```

```python
# Task 1.4
def hash_search(tbl, addr):
    k = hash(addr)
    item = tbl[k]

    if item == addr:
        return True

    found = False
    index = k
    while not found:
        index = (index + 1) % len(tbl)
        if index == k:
            break

        item = tbl[index]

        if item == addr:
            found = True

    return found

print(hash_search(hash_table, '117,Jurong East Avenue 1,S228488'))
print(hash_search(hash_table, 'meowmoeeea'))
```

```python
# Task 2.1
# <A>: lo <- lo + 1
# <B>: hi <- hi - 1
# <C>: temp <- seq[start]
#      seq[start] <- seq[hi]
#      seq[hi] <- temp

def quicksort(lst):
    quicksort_helper(lst, 0, len(lst) - 1)
    return lst

def quicksort_helper(lst, start, end):
    if start < end:
        mid = partition(lst, start, end)
        quicksort_helper(lst, start, mid - 1)
        quicksort_helper(lst, mid + 1, end)

    return lst

def partition(lst, start, end):
    pivot = lst[start]
    lo = start + 1
    hi = end
    done = False

    while done == False:
        while lo <= hi and lst[lo] <= pivot:
            lo += 1

        while lst[hi] >= pivot and hi >= lo:
            hi -= 1

        if hi < lo:
            done = True
        else:
            temp = lst[lo]
            lst[lo] = lst[hi]
            lst[hi] = temp

    lst[start], lst[hi] = lst[hi], lst[start]
    return hi
```

```python
# Task 2.3
def insertion_sort(lst):
    n = len(lst)

    for i in range(1, n):
        key = lst[i]
        j = i - 1

        while j >= 0 and lst[j] > key:
            lst[j + 1] = lst[j]
            j -= 1

        lst[j + 1] = key

    return lst

with open('./TASK2_3_Akshat.TXT', 'w') as f:
    f.writelines([str(i) +'\n' for i in insertion_sort(data)])
```

```python
# Task 3.4
def insert0(array, col):
    column = [i[col] for i in array]
    if '' not in column:
        print("Player (O) missed a turn.")
        return

    chosen_index = 0
    while chosen_index < len(array) and column[chosen_index] == '':
        chosen_index += 1
    chosen_index -= 1

    array[chosen_index][col] = 'O'

    return array

print(*insert0(grid1, 2), sep='\n')
```

```python
# Task 3.5
def insertX(array, col):
    column = [i[col] for i in array]
    if '' not in column:
        print("Player (X) missed a turn.")
        return

    chosen_index = 0
    while chosen_index < len(array) and column[chosen_index] == '':
        chosen_index += 1
    chosen_index -= 1

    array[chosen_index][col] = 'X'

    return array
```

```python
# Task 3.6
def checkrow(array):
    for row in array:
        consecutive = 0
        prev = None

        for char in row:
            if char == '':
                prev = None
                consecutive = 0
                continue

            if char != prev and char != '':
                prev = char
                consecutive = 0

            if char == prev:
                consecutive += 1

            if consecutive == 4:
                print(row)
                return True
    return False
```

```python
# Task 3.7
def checkcolumn(array):
    columns = list(zip(*array))
    return checkrow(columns)

checkcolumn(grid1)
```

```python
def checkdiagonal(array):
    for row in range(3):
        for col in range(4):
            
            diagonal1 = []
            diagonal2 = []
            for i in range(4):
                diagonal1.append(array[row+i][col+i])
                diagonal2.append(array[row+3-i][col+i])

            string = ''.join(diagonal1)
            if 'OOOO' in string or 'XXXX' in string:
                return True
            else:
                string = ''.join(diagonal2)
                if 'OOOO' in string or 'XXXX' in string:
                    return True
    return False
```

```python
# Task 5.1
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

    def get_data(self):
        return self.data
    def set_data(self, data):
        self.data = data

    def get_next(self):
        return self.next
    def set_next(self, next):
        self.next = next
```

```python
# Task 5.2
from datetime import *

def convert(string):              #string: date in 'yyyy/mm/dd' format
    return datetime.strptime(string, '%Y/%m/%d').date()

def expirydate(string):           #string: date in 'yyyy/mm/dd' format
    now = convert(string)
    four_weeks = timedelta(weeks=4)
    later = now + four_weeks
    return later

class Queue:
    def __init__(self):
        self.first = None

    def enqueue(self, new_data):
        if not self.first:
            self.first = Node(new_data)
            return

        current = self.first
        while current.next:
            current = current.next

        current.next = Node(new_data)

    def dequeue(self):
        if not self.first:
            return

        data = self.first.get_data()
        self.first = self.first.next
        return data

    def peek(self):
        return self.first.get_data()

    def display(self):
        if not self.first:
            print('Empty!')
            return

        current = self.first
        while current:
            print(current.get_data(), end='')
            if current.next:
                print('->', end='')
            
            current = current.next
        print()

    def update(self, string):
        if not self.first:
            return

        date = convert(string)
        previous = None
        current = self.first
        while current:
            expiry = expirydate(current.data[1])
            if (date-expiry).days > 7*4:
                if previous:
                    previous.next = current.next
                else:
                    self.first = current.next
            else:
                previous = current
            current = current.next

    def use(self, n):
        if not self.first:
            return

        items = []
        current = self.first
        while current:
            items.append(current)
            current = current.next

        items.sort(key=lambda x: expirydate(x.data[1]))
        if sum([i[0] for i in items]) > n:
            print(f'There is insufficient reward points, {n} points not deducted')
            return

        
```
