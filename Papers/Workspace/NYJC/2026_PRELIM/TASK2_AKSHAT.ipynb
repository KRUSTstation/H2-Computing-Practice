```python
# Task 2.1
import GAME_LOGIC as g

from socket import socket

s = socket()
s.bind(('localhost', 65000))

s.listen()

c, addr = s.accept()

c.send('WELCOME'.encode())
c.send('ENTER NAME'.encode())

name = c.recv(4096).decode()
game = g.Game(name)
game.generate_items()

c.send(game.player.status().encode())
while game.player.turns_taken < game.max_turns:
    c.send('YOUR TURN'.encode())
    
```
