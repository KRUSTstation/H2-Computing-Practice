```python
# Task 2.1
class Node:
    def __init__(self):
        self.data = ''
        self.pointer = -1

    def get_data(self):
        return self.data
    def set_data(self, data):
        self.data = data

    def get_pointer(self):
        return self.pointer
    def set_pointer(self, data):
        self.pointer = data
```

```python
# Task 2.2
class SFSS:
    def __init__(self, n: int):
        self.array = [Node() for _ in range(n)]
        self.directory = {}
        self.free = 0
        for i, item in enumerate(self.array):
            if i != n-1:
                item.set_pointer(i+1)

    def get_free(self):
        return self.free
    def set_free(self, data):
        self.free = data

    def get_array(self):
        return self.array

    def get_directory(self):
        return self.directory

    def save_file(self, filename, datastring):
        if len(self.array) <= 0:
            print('Array is empty!')
            return

        self.directory[filename] = self.free

        for i, char in enumerate(datastring):
            self.array[self.free + i].set_data(char)
            if i == len(datastring) - 1:
                self.array[i].set_pointer(-1)
                self.free = i+1

    def delete_file(self, filename):
        if self.directory.get(filename) == None:
            print('File does not exist')
            return

        starting_i = self.directory[filename]
        self.directory[filename] = None
        
        current = self.array[starting_i]
        
        current.set_data('')
        #current.set_pointer(self.free)

        count = 0
        while current.get_pointer() != -1:
            print(current.get_pointer())
            print(current.get_data())
            current = self.array[current.get_pointer()]
            current.set_data('')
            current.set_pointer(count)
            count += 1

        self.free = count + 1

    def display(self):
        print(f'File Directory: {self.directory}')
        print(f'Free space pointer: {self.free}')
        print(f'Array State: ')

        print(f"Index | Data | Next")
        for i, item in enumerate(self.array):
            print(f'{i:<5} | {item.get_data():<5} | {item.get_pointer():<5}')
```

```python
# Task 2.3
storage = SFSS(10)
storage.save_file('file1', 'ace')
#storage.display()
storage.save_file('file2', 'man')
storage.display()
storage.delete_file('file1')
storage.display()
storage.save_file('file3', 'nudhigh')
storage.display()
```
