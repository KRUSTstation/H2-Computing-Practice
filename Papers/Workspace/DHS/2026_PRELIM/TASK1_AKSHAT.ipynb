```python
# Task 1.1
def load_key(filename):
    with open(filename) as f:
        n1, n2 = [int(i.replace(' ', '')) for i in f.read().split(',')]

    return n1, n2

load_key('./Resources/publickey.txt')
```

```python
# Task 1.2
def load_message(filename):
    with open(filename) as f:
        return f.read()

load_message('./Resources/plaintext.txt')
```

```python
# Task 1.3
def encrypt_message(message, e , n):
    encrypted = []
    for char in message:
        encrypted.append(ord(char)**e % n)

    return encrypted

encrypt_message(load_message('./Resources/plaintext.txt'), *load_key('./Resources/publickey.txt'))
```

```python
# Task 1.4
def decrypt_message(integers, d, n):
    decrypted = ''
    for item in integers:
        decrypted += chr(item**d % n)

    return decrypted

#decrypt_message()
```

```python
# Task 1.5
pub, pri = load_key('./Resources/publickey.txt'), load_key('./Resources/privatekey.txt')
print(pub, pri)

msg = load_message('./Resources/plaintext.txt')
print(msg)

encrypted = encrypt_message(msg, *pub)
print(encrypted)

decrypted = decrypt_message(encrypted, *pri)
print(decrypted)
```
