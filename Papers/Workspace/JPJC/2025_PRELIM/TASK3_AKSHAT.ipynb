```python
# Task 3.1
def encrypt(code):
    encrypted = ''

    for let in code:
        encrypted += chr(ord('a') + (ord(let) + 3) % (ord('z')+1))

    return encrypted

encrypt('z')
```

```python
# Task 3.2
def hashFunction(locker_id):
    return int(locker_id) % 37
```

```python
# Task 3.3
def add(locker_id, hta):
    code = input('Enter code word')
    
    if code.isalpha() and code.islower():
        encrypted = encrypt(code)

        index = hashFunction(locker_id)

        while hta[index] != None:
            index = (index + 1) % len(hta)

        hta[index] = (str(locker_id), encrypted)
```

```python
# Task 3.4
def search(hta):
    locker_id = input('Enter locker id')
    code = input('Enter the codeword')

    if code.isalpha() and code.islower():
        encrypted = encrypt(code)

        index = hashFunction(locker_id)

        while hta[index] != None:
            if hta[index][0] == locker_id:
                if hta[index][1] == encrypted:
                    print('LOCKER IS OPEN')
                else:
                    print('INCORRECT CODE')

                return
            
            index = (index + 1) % len(hta)

        print('LOCKER NOT FOUND')
```

```python
# Task 3.5
hta = [None for _ in range(37)]

add(36, hta)
add(73, hta)

search(hta)
search(hta)
```
