```python
# Task 2.1
class PageNode:
    def __init__(self, url):
        self.url = url
        self.next = None

class BrowserHistory:
    def __init__(self):
        self.head = None

    def visit_page(self, url):
        if not self.head:
            self.head = PageNode(url)
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = PageNode(url)

    def print_history(self):
        if not self.head:
            print('Empty')
            return

        current = self.head
        while current:
            print(current.url)
            current = current.next
```

```python
# Task 2.2
class BrowserNavigator:
    def __init__(self):
        self.current_page = None
        self.back_stack = [None for _ in range(10)]
        self.forward_stack = [None for _ in range(10)]
        self.back_top = 0
        self.forward_top = 0

    def go_back(self):
        if self.forward_top >= 10 or self.back_top < 0: return

        self.forward_stack[self.forward_top] = self.current_page
        self.current_page = self.back_stack[self.back_top-1]
        self.back_stack[self.back_top] = None

        self.back_top -= 1
        self.forward_top += 1

    def go_forward(self):
        if self.back_top >= 10 or self.forward_top < 0: return

        self.back_stack[self.back_top] = self.current_page
        self.current_page = self.forward_stack[self.forward_top-1]
        self.forward_stack[self.forward_top] = None

        self.back_top += 1
        self.forward_top -= 1

    def visit_new_page(self, url):
        if self.back_top >= 10: return

        self.back_stack[self.back_top] = self.current_page
        self.current_page = url
        self.forward_stack = [None for _ in range(10)]

        self.back_top += 1
        self.forward_top = 0
```

```python
# Task 2.3
BH = BrowserHistory()
BN = BrowserNavigator()

BH.visit_page('jpjc.coursemology.org')
BN.visit_new_page('jpjc.coursemology.org')

BH.visit_page('jpjc.moe.edu.sg')
BN.visit_new_page('jpjc.moe.edu.sg')

BH.visit_page('web.whatsapp.com')
BN.visit_new_page('web.whatsapp.com')

BN.go_back()
BH.visit_page(BN.current_page)
BN.go_back()
BH.visit_page(BN.current_page)

BN.go_forward()
BH.visit_page(BN.current_page)

print(BN.current_page)
print()
BH.print_history()
```
