```python
# Task 1.1
import csv

def readData(filename):
    with open(filename) as f:
        data = [[int(j) if j.isdigit() else j for j in i] for i in csv.reader(f)][1:]

    return data

readData('TASK1.csv')
```

```python
# Task 1.2
def quicksort(treasures):
    if len(treasures) <= 1:
        return treasures

    pivot = treasures[0][0]

    low = [i for i in treasures if i[0] < pivot]
    equal = [i for i in treasures if i[0] == pivot]
    more = [i for i in treasures if i[0] > pivot]

    return quicksort(low) + equal + quicksort(more)

quicksort(readData('TASK1.csv'))
```

```python
# Task 1.3
def binarySearch(treasures, name):
    high = len(name)
    low = 0

    while low <= high:
        mid = (low + high) // 2

        if treasures[mid][0] == name:
            return treasures[mid][1]

        elif name < treasures[mid][0]:
            high = mid - 1

        else:
            low = mid + 1

    return -1

binarySearch(quicksort(readData('TASK1.csv')), 'Dragon Scale')
```

```python
# Task 1.4
def menu():
    print("""
    Welcome to Your Treasure Inventory!
    1. View Inventory
    2. Sort Treasures by Name
    3. Search Treasure by Name
    4. Exit\n
    """)

data = readData('TASK1.csv')

user = None

while user != '4':
    menu()
    user = input('Choose option: ')

    if user == '1':
        for item in data:
            print(f'{item[0]} (Value: {item[1]})')

    elif user == '2':
        data = quicksort(data)

        for item in data:
            print(f'{item[0]} (Value: {item[1]})')

    elif user == '3':
        name = input('Enter treasure name to search: ')

        val = binarySearch(data, name)
        if val == -1:
            print('Oops! Nothing found!')
        else:
            print(f"Found '{name}' worth {val} gold coins!")

    elif user != '4':
        print('INVALID')

print('Goodbye')
```
