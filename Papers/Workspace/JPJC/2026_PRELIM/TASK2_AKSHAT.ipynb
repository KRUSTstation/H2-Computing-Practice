```python
# Task 2.1
class Tile:
    def __init__(self, symbol):
        self.symbol = symbol

    def get_symbol(self):
        return self.symbol

class Treasure(Tile):
    def __init__(self, symbol, treasure):
        super().__init__(symbol)
        self.treasure = treasure

    def get_treasure(self):
        return self.treasure

class Bomb(Tile):
    def __init__(self, symbol, damage):
        super().__init__(symbol)
        self.damage = damage

    def get_damage(self):
        return self.damage
```

```python
# Task 2.2
def display_grid(grid):
    for row in grid:
        for item in row:
            print(item.get_symbol(), end=' ')
        print()
```

```python
# Task 2.3
def main():
    grid = [[Tile('.') for _ in range(5)] for _ in range(5)]

    grid[1][2] = Treasure('T', 'Silver key')
    grid[2][4] = Treasure('T', 'Gold coin')
    grid[3][3] = Treasure('T', 'Jade pendant')

    grid[2][3] = Bomb('B', 1)
    grid[0][0] = Tile('P')

    display_grid(grid)

    lives = 1
    pos = [0, 0]
    inventory = []

    treasures_left = 3

    def move(key, position):
        if key == 'W':
            value = -1
            index = 0

        elif key == 'S':
            value = 1
            index = 0

        elif key == 'A':
            value = -1
            index = 1

        elif key == 'D':
            value = 1
            index = 1

        old_pos = [i for i in position]

        position[index] += value
        
        for i in position:
            if i < 0:
                return old_pos

        return position

    moves = input('Enter sequence of moves: ').upper()

    for direction in moves:
        if lives <= 0:
            print('Game over no more lives')
            break
        
        pos = move(direction, pos)

        item = grid[pos[0]][pos[1]]
        
        if item.get_symbol() == 'B':
            lives -= item.get_damage()
        elif item.get_symbol() == 'T':
            treasure = item.get_treasure()

            if treasure not in inventory:
                inventory.append(treasure)
                treasures_left -= 1
                
    if lives > 0:
        if treasures_left > 0:
            print('Sorry you did not win the game')
        else:
            print('Congratulations! You won the game!')
    
        print(f'Treasures collected {3 - treasures_left}/3: ', end = '')
        for item in inventory:
            print(f'{item}, ', end='')
        print()

main()
main()
main()
```
