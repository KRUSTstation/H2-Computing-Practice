```python
# Task 1.1
import csv

def task1_1(filename):
    with open(f'./data files/{filename}') as f:
        data = [(int(i[0]), i[1], float(i[2]), int(i[3])) for i in csv.reader(f)]

    return data

result = task1_1("TASK1.txt")
print(result)
```

```python
# Task 1.2
def task1_2(list_of_records):
    for i in range(len(list_of_records)):
        for j in range(len(list_of_records)):
            if list_of_records[j][2] > list_of_records[i][2]:
                list_of_records[i], list_of_records[j] = list_of_records[j], list_of_records[i]

    return list_of_records

print(task1_2(task1_1("TASK1.txt")))
```

```python
# Task 1.3
def task1_3(list_of_records):
    if len(list_of_records) <= 1:
        return list_of_records

    pivot = list_of_records[len(list_of_records) // 2][0]

    less = [i for i in list_of_records if i[0] < pivot]
    equal = [i for i in list_of_records if i[0] == pivot]
    more = [i for i in list_of_records if i[0] > pivot]

    return task1_3(less) + equal + task1_3(more)


print(task1_3(task1_1("TASK1.txt")))
```

```python
# Task 1.4
def bin_search(list_of_tuples, prod_id, low=None, high=None):
    if low == None or high == None:
        low = 0
        high = len(list_of_tuples) - 1

    if low > high:
        return "Not Found"

    mid = (low + high) // 2

    if prod_id < list_of_tuples[mid][0]:
        return bin_search(list_of_tuples, prod_id, low, mid - 1)

    elif prod_id > list_of_tuples[mid][0]:
        return bin_search(list_of_tuples, prod_id, mid+1, high)

    return list_of_tuples[mid][1:]

def main():
    prod_id = int(input("Input product ID: "))
    result = bin_search(task1_3(task1_1("TASK1.txt")), prod_id)
    print(result)

main()
main()
```
