```python
# Task 3.1
def createSensorNode():
    from random import randint

    sensorID = str(randint(0, 99999))

    while len(sensorID) < 5:
        sensorID = '0' + sensorID

    return {"Left": None, "SensorID": sensorID, "Right": None}

print(createSensorNode())
```

```python
# Task 3.2a
def insert(Root, newSensorNode):
    if Root == None:
        return newSensorNode

    previous = Root
    current = Root['Left'] if int(newSensorNode['SensorID']) < int(Root['SensorID']) else Root['Right']
    while current:
        previous = current
        
        if int(newSensorNode['SensorID']) < int(current['SensorID']):
            current = current['Left']
        else:
            current = current['Right']

    if int(newSensorNode['SensorID']) < int(previous['SensorID']):
        previous['Left'] = newSensorNode
    else:
        previous['Right'] = newSensorNode

    return Root
```

```python
# Task 3.2b
def insertHandleDuplicate(Root, newSensorNode):
    if Root == None:
        return newSensorNode

    inserted = False
        
    while inserted == False:
        isRepeat = False
        
        if int(newSensorNode['SensorID']) == int(Root['SensorID']):
            newSensorNode = createSensorNode()
            continue
        
        previous = Root
        current = Root['Left'] if int(newSensorNode['SensorID']) < int(Root['SensorID']) else Root['Right']
        while current:
            previous = current
            
            if int(newSensorNode['SensorID']) < int(current['SensorID']):
                current = current['Left']
            elif int(newSensorNode['SensorID']) > int(current['SensorID']):
                current = current['Right']
            else:
                isRepeat = True
                break

        if isRepeat: continue
    
        if int(newSensorNode['SensorID']) < int(previous['SensorID']):
            previous['Left'] = newSensorNode
        elif int(newSensorNode['SensorID']) > int(previous['SensorID']):
            previous['Right'] = newSensorNode
        else:
            continue

        inserted = True

    return Root
```

```python
# Task 3.3
def inOrder(Root, result=None):
    if result is None:
        result = []

    if Root:
        inOrder(Root['Left'], result)
        result.append(Root['SensorID'])
        inOrder(Root['Right'], result)

    return result
```

```python
# Task 3.4
TRoot = None

for _ in range(10):
    newSensorNode = createSensorNode()
    TRoot = insert(TRoot, newSensorNode)

print(inOrder(TRoot))
```

```python
# Task 3.5
def isValidSensorID(sensorID):
    return len(sensorID) == 5 and 0 <= int(sensorID) <= 99999
```

```python
# Task 3.6
isValidSensorID('12345') # normal since it is a reasonable value inside the range of 0 and 99999 and clearly has a  length of 5
# returns True

isValidSensorID('00000') # Boundary since its at the edge of the range at 0
# returns True

isValidSensorID('123') # Erroneous since its length is not 5
# returns False
```
