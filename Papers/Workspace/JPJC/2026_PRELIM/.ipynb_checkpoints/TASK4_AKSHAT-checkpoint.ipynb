```python
# Task 4.1
import sqlite3

conn = sqlite3.connect('HOUSEALLOCATION.db')
cursor = conn.cursor()

with conn:
    cursor.executescript('''
        DROP TABLE IF EXISTS HOUSE;
        DROP TABLE IF EXISTS STAFF;
    
        CREATE TABLE IF NOT EXISTS HOUSE(
            houseName TEXT PRIMARY KEY,
            imageURL TEXT NOT NULL
        );

        CREATE TABLE IF NOT EXISTS STAFF(
            staffID TEXT PRIMARY KEY,
            name TEXT NOT NULL,
            department TEXT NOT NULL,
            houseName TEXT NOT NULL,

            FOREIGN KEY (houseName) REFERENCES HOUSE(housename)
        );
    ''')
    
    conn.commit()
```

```python
# Task 4.2
import csv
import sqlite3

conn = sqlite3.connect('HOUSEALLOCATION.db')
cursor = conn.cursor()

with conn:
    with open('./data files/STAFF.txt') as f:
        data = [i for i in csv.reader(f)]

        for item in data:
            cursor.execute('INSERT INTO STAFF VALUES (?, ?, ?, ?)', item)

    with open('./data files/HOUSE.txt') as f:
        data = [i for i in csv.reader(f)]

        for item in data:
            cursor.execute('INSERT INTO HOUSE VALUES (?, ?)', item)
    
    conn.commit()
```
