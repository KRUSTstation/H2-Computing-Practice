```python
# Task 3.1
def task3_1():
    alphabet = [chr(65 + i) for i in range(26)]

    table = []

    for i in range(26):
        row = [alphabet[i]]
        count = (i + 1) % 26
        while alphabet[count] != alphabet[i]:
            row.append(alphabet[count])
            count = (count + 1) % 26
        
        table.append(row)

    return table

for i in task3_1():
    print(i)
```

```python
# Task 3.2
def task3_2(keyword, plaintext, vtable):
    alphabet = [chr(65 + i) for i in range(26)]

    new_pt = ''
    for char in plaintext:
        if char.isalnum():
            new_pt += char.upper()

    encrypted = ''
    count = 0
    while count < len(new_pt):
        pt_char = new_pt[count]
        key_char = keyword[count % len(keyword)]

        encrypted += vtable[alphabet.index(pt_char)][alphabet.index(key_char)]
        if (count+1) % 5 == 0:
            encrypted += ' '

        count += 1

    return encrypted

task3_2('PYTHON', 'HELLO WORLD WRITTEN IN JAVA', task3_1())
```
