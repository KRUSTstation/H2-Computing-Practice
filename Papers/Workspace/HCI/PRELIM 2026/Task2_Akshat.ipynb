```python
# Task 2.2
class Playlist:
    def __init__(self):
        self.head = None

    def isEmpty(self):
        return self.head == None

    def addSong(self, node):
        if self.isEmpty():
            self.head = node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = node

    def playNext(self):
        if self.isEmpty():
            return

        head = self.head
        self.head = self.head.next

        head.next = None

        return head

    def displayPlaylist(self):
        if self.isEmpty():
            print('[]')

        print('[', end='')
        current = self.head
        while current.next:
            print(current.title + ', ', end='')
            current = current.next

        print(current.title + ']')
```

```python
# Task 2.4
class Playlist:
    def __init__(self):
        self.head = None

    def isEmpty(self):
        return self.head == None

    def addSong(self, node):
        if self.isEmpty():
            self.head = node
            return

        current = self.head
        while current.next:
            current = current.next

        current.next = node

    def playNext(self):
        if self.isEmpty():
            return

        head = self.head
        self.head = self.head.next

        head.next = None

        return head

    def search(self, title):
        if self.isEmpty():
            return

        current = self.head
        while current:
            if current.title == title:
                return current

            current = current.next

        return

    def delete(self, title):
        if self.isEmpty() or self.search(title) == None:
            return False

        if self.head.title == title:
            old_head = self.head
            self.head = self.head.next

            old_head.next = None
            return True

        current = self.head.next
        previous = self.head
        while current:
            if current.title == title:
                old_current = current
                previous.next = current.next

                old_current.next = None

                return True

            previous = current
            current = current.next

        return False

    def insert(self, node, position):
        if position <= 1:
            node.next = self.head
            self.head = node
            return

        count = 2
        current = self.head.next
        previous = self.head
        while current:
            if count == position:
                previous.next = node
                node.next = current
                return

            previous = current
            current = current.next
            count += 1

        previous.next = node

    def displayPlaylist(self):
        if self.isEmpty():
            print('[]')

        print('[', end='')
        current = self.head
        while current.next:
            print(current.title + ', ', end='')
            current = current.next

        print(current.title + ']')
```

```python
import csv

my_playlist = Playlist()
with open('PLAYLIST.txt') as f:
    for song in [i for i in csv.reader(f)]:
        my_playlist.addSong(SongNode(*song))

print('Search result for Silver')
print(my_playlist.search('Silver'))

my_playlist.delete("Mom's Pasta")

print('\nAFTER DELETING MOMS PASTA')
my_playlist.displayPlaylist()

my_playlist.insert(SongNode('Smile Till the End', 'Ms Gugu & Uno Cars'), 3)

print('\nAFTER INSERTING NEW SONG')
my_playlist.displayPlaylist()
```
