```python
# Task 1.1
import csv

def read_file(filename):
    with open(filename) as f:
        data = [[int(j) if j.isdigit() else str(j) for j in i] for i in csv.reader(f)]

    return data

print(read_file('song.txt'))
```

```python
# Task 1.2
def binary_search(song_list, new_length):
    start = 0
    end = len(song_list)
    found = []
    found_index = -1

    while start < end:
        mid = (start + end) // 2

        if song_list[mid][0] == new_length:
            found.append(song_list[mid][2])
            found_index = mid
            break

        if new_length < song_list[mid][0]:
            end = mid - 1
        else:
            start = mid + 1

    if found_index != -1:
        # go left
        i = found_index - 1
        while i >= 0 and song_list[i][0] == new_length:
            found.append(song_list[mid][2])
            i -= 1

        # right
        i = found_index + 1
        while i < len(song_list) and song_list[i][0] == new_length:
            found.append(song_list[mid][2])
            i += 1

    return found

print(binary_search(read_file('song.txt'), 240) == [])
print(binary_search(read_file('song.txt'), 260) == ['Rock', 'Rock'])
```

```python
# Task 1.3
def distance(song_list, new_length, new_tempo):
    for song in song_list:
        length = song[0]
        tempo = song[1]

        dist = ((new_length - length) ** 2 + (new_tempo - tempo) ** 2) ** 0.5
        song.append(dist)

    return song_list

print(distance(read_file('song.txt'), 280, 115))
```

```python
# Task 1.4
def quicksort(arr):
    if len(arr) <= 1:
        return arr

    pivot = len(arr) // 2
    mid = arr[pivot][3]

    equal = [i for i in arr if i[3] == mid]
    less = [i for i in arr if i[3] < mid]
    more = [i for i in arr if i[3] > mid]

    return (quicksort(less) + equal + quicksort(more))

def task1_4(arr):
    s = quicksort(arr)
    return [i[2] for i in s[:5]]

print(task1_4(distance(read_file('song.txt'), 280, 115)))
```

```python
# Task 1.5
def task1_5(arr, length, tempo, genre):
    arr.append([length, tempo, genre])
    s = insertion_sort(arr)

    with open('new_song.txt', 'w') as f:
        for song in s:
            for item in song:
                f.write(str(item) + ',')
            f.write('\n')

def insertion_sort(arr):
    for i in range(1, len(arr)):
        while i > 0 and arr[i][0] < arr[i-1][0]:
            arr[i], arr[i-1] = arr[i-1], arr[i]
            i -= 1

    for i in range(1, len(arr)):
        while i > 0 and arr[i][1] < arr[i-1][1]:
            arr[i], arr[i-1] = arr[i-1], arr[i]
            i -= 1

    return arr

task1_5(read_file('song.txt'), 310, 120, 'Classical')
```
