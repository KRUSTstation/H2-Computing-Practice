```python
# Task 1.1
def task1_1(key, message):
    if len(key) == 0 or len(message) == 0:
        return ''

    message = message.replace(' ', 'z')
    grid = [[i for i in key]] 
    for _ in range(int(len(message)/len(key)) + 1):
        grid.append([None for _ in key])

    count = 0 
    for item in grid[1:]:
        for i in range(len(item)):
            if count >= len(message):
                item[i] = 'x'
                continue

            item[i] = message[count]

            count+=1

    # print(grid)
    sorted_key = sorted(grid[0], key=lambda x: ord(x))

    result = ""
    used_chars = []
    for char in sorted_key:
        if char in used_chars:
            item = grid[grid[0].index(char) + 1]
            for c in grid[1:]:
                result += c[item]
        else:
            item = grid[0].index(char)
            for c in grid[1:]:
                result += c[item]
        
        used_chars.append(char)

    return result
```

```python
# Task 1.2
print(task1_1('ALGORITHM', 'computing is an interesting subject'))
```
