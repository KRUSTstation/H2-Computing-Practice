```python
# Task 3.1
class TaskNode:
    def __init__(self, taskName, priority):
        self.taskName = taskName
        self.priority = priority
        self.next = None

    def __str__(self):
        return f'({self.taskName}, {self.priority})'
```

```python
# Task 3.2
class PriorityQueue:
    def __init__(self):
        self.first = None

    def viewPendingTasks(self):
        current = self.first
        while current:
            print(current)
            current = current.next

    def enqueue(self, taskNode):
        if not self.first:
            self.first = taskNode
            return

        current = self.first
        previous = None
        while current:
            if taskNode.priority > current.priority:
                if previous:
                    previous.next = taskNode
                taskNode.next = current

                if self.first == current:
                    self.first = taskNode

                return

            previous = current
            current = current.next

        previous.next = taskNode

    def isEmpty(self):
        return self.first == None

    def dequeue(self):
        if self.isEmpty():
            return

        first = self.first
        second = self.first.next
        self.first = second

        return first

    def delete(self, taskName):
        if not self.first:
            return False

        current = self.first
        previous = None
        while current:
            if current.taskName == taskName:
                if current == self.first:
                    self.first = current.next
                else:
                    previous.next = current.next

                #current.next = None
                return True

            previous = current
            current = current.next

        return False

    def search(self, taskName):
        if self.isEmpty():
            return

        current = self.first
        while current:
            if current.taskName == taskName:
                return current

            current = current.next

        return

    def updatePriority(self, taskName, newPriority):
        item = TaskNode(taskName, newPriority)
        if self.delete(taskName):
            self.enqueue(item)
```

```python
q = PriorityQueue()
with open('./TASKS.txt') as f:
    data = [i.strip().split(',') for i in f.readlines()]

for i in data:
    q.enqueue(TaskNode(i[0], int(i[1])))

print('Initial queue')
q.viewPendingTasks()

q.dequeue()
q.dequeue()

print('After dequeueing:')
q.viewPendingTasks()
```
