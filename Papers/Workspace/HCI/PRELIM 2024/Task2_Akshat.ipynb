```python
# Task 2.1
def read_csv(filename):
    import csv

    with open(filename) as f:
        data = [[int(j) if j.isdigit() else j for j in i] for i in csv.reader(f)]

    return data

colours = read_csv("paint.csv")
print(len(colours))
print(colours)
```

```python
# Task 2.2
def find(array, r, g, b):
    for item in array:
        if item[0] == r and item[1] == g and item[2] == b:
            return item[3]

    return 'nil'

print(find(colours, 253, 253, 5))
print(find(colours, 5, 253, 253))
```

```python
# Task 2.3
def bubble(arr, color):
    d = {'red': 0, 'green': 1, 'blue': 2}
    index = d[color]

    for i in range(len(arr)-1):
        for j in range(len(arr)-i-1):
            if arr[j][index] < arr[j+1][index]:
                arr[j], arr[j+1] = arr[j+1], arr[j]

    return arr

def rgb(array, colour):
    if colour not in ['red', 'green', 'blue']:
        print('Invalid colour')
        return []

    return bubble(array, colour)

print(rgb(colours, 'yellow'))
print(rgb(colours, 'blue'))
```

```python
# Task 2.4
def quicksort(arr):
    if len(arr) <= 1:
        return arr

    pivot = len(arr) // 2
    mid = sum(arr[pivot][0:3])

    equal = [i for i in arr if mid == sum(i[0:3])]
    more = [i for i in arr if mid < sum(i[0:3])]
    less = [i for i in arr if mid > sum(i[0:3])]

    return quicksort(more) + equal + quicksort(less)

def bright(array):
    return quicksort(array)

print(bright(colours))
```

```python
# Task 2.5
def menu():
    colours = read_csv('paint.csv')
    print('What you wan (red/green/blue/dark/bright): ')
    user = None
    while user not in ['red', 'green', 'blue', 'dark', 'bright']:
        user = input().lower()

    if user in ['red', 'green', 'blue']:
        arr = rgb(colours, user)
    elif user == 'bright':
        arr = bright(colours)
    else:
        arr = list(reversed(bright(colours)))

    print(arr[:3])

menu()
menu()
menu()
```
