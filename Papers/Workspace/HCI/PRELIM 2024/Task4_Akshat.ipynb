```python
# Task 4.1
import sqlite3

conn = sqlite3.connect('TRIP.db')
cursor = conn.cursor()

with conn:
    cursor.executescript('''
        DROP TABLE Flight;
        DROP TABLE Customer;
        DROP TABLE Ticket;

        CREATE TABLE IF NOT EXISTS Flight(
            FlightNo TEXT PRIMARY KEY,
            DepartCity TEXT NOT NULL,
            ArrivalCity TEXT NOT NULL,
            DepartTime TEXT NOT NULL,
            ArrivalTime TEXT NOT NULL
        );

        CREATE TABLE IF NOT EXISTS Customer(
            CustomerNo INTEGER PRIMARY KEY,
            Name TEXT NOT NULL,
            Gender TEXT NOT NULL,
            DOB INT NOT NULL
        );

        CREATE TABLE IF NOT EXISTS Ticket(
            TicketNo INTEGER PRIMARY KEY,
            CustomerNo INT NOT NULL,
            FlightNo TEXT NOT NULL,
            Date TEXT NOT NULL,
            Seat TEXT NOT NULL,

            FOREIGN KEY (CustomerNo) REFERENCES Customer(CustomerNo),
            FOREIGN KEY (FlightNo) REFERENCES Flight(FlightNo)
        );
    ''')

    conn.commit()
```

```python
# Task 4.2
import csv
import sqlite3

conn = sqlite3.connect('TRIP.db')
cursor = conn.cursor()

with conn:
    with open('./FLIGHT.txt') as f:
        data = list(csv.reader(f))
        #data = [[int(i) if i.isdigit() else i for i in item] for item in csv.reader(f)]
        for i in data:
            cursor.execute('INSERT INTO Flight(FlightNo, DepartCity, ArrivalCity, DepartTime, ArrivalTime) VALUES (?, ?, ?, ?, ?)', i)

    with open('./CUSTOMER.txt') as f:
        data = list(csv.reader(f))
        for i in data:
            cursor.execute('INSERT INTO Customer(CustomerNo, Name, Gender, DOB) VALUES (?, ?, ?, ?)', i)

    with open('./TICKET.txt') as f:
        data = list(csv.reader(f))
        for i in data:
            cursor.execute('INSERT INTO Ticket(TicketNo, CustomerNo, FlightNo, Date, Seat) VALUES (?, ?, ?, ?, ?)', i)

    conn.commit()
    
```
