```python
# Task 1.1
def task1_1(artefacts_list):
    valid = []
    
    for artefact in artefacts_list:
        ID = artefact[0]
        
        if ID[0].isalpha() and ID[0].isupper() and ID[1:len(ID)].isdigit():
            num = int(ID[1:len(ID)])
            if 1000 <= num <= 9999:
                valid.append(artefact)

    return valid

v = task1_1([("A1234", 85),("B1234", 85),("C1234", 85),("B2567", 92),("A3891", 45),("C4012", 78),("B5678", 95),("D6543", 88),("E7890", 65),("A8123", 73),("C9012", 98),("B1567", 52)])
print(v[:5])
```

```python
# Task 1.2
def task1_2(artefacts_list):
    if len(artefacts_list) <= 1:
        return artefacts_list

    pivot = artefacts_list[0][1]

    less = [i for i in artefacts_list if i[1] < pivot]
    equal = [i for i in artefacts_list if i[1] == pivot]
    more = [i for i in artefacts_list if i[1] > pivot]

    return task1_2(more) + equal + task1_2(less)

s = task1_2([("A1234", 85),("B1234", 85),("C1234", 85),("B2567", 92),("A3891", 45),("C4012", 78),("B5678", 95),("D6543", 88),("E7890", 65),("A8123", 73),("C9012", 98),("B1567", 52)])
print(s)
```

```python
# Task 1.3
def binary_search(arr, target, low, high):
    if low > high:
        return
    
    mid = (low + high) // 2

    if target < arr[mid][1]:
        return binary_search(arr, target, mid+1, high)
        
    elif target == arr[mid][1]:
        return arr[mid]

    else:
        return binary_search(arr, target, low, mid-1)

def task1_3(artefacts_list, target_rating):
    s = task1_2(artefacts_list)
    return binary_search(s, target_rating, 0, len(s)-1)

L = [("A1234", 85),("B1234", 85),("C1234", 85),("B2567", 92),("A3891", 45),("C4012", 78),("B5678", 95),("D6543", 88),("E7890", 65),("A8123", 73),("C9012", 98),("B1567", 52)]

print('Finding rating 85')
found = task1_3(L, 85)
if found == None:
    print('Not Found')
else:
    print(found)

print('Finding rating 100')
found = task1_3(L, 100)
if found == None:
    print('Not Found')
else:
    print(found)
```
