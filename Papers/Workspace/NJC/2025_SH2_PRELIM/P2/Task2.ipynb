```python
#Task 2.1
import csv

def read_file(filename):
    with open(filename) as f:
        data = [tuple(i) for i in csv.reader(f)]
    
    return data
```

```python
#Task 2.2

def get_roles(candidates):
    roles = {}

    for info in candidates:
        if info[1] in roles.keys():
            roles[info[1]].append(info)
        else:
            roles[info[1]] = [info]

    return roles
```

```python
#Task 2.3

roles = get_roles(read_file('./Resources/candidates.txt'))

print(f"{'Roles':<15}{'Count':<8}{'Low Score':<8}{'High Score':<8}")
for role, info in roles.items():
    print(f"{role:<15}{len(info):<8}{min(info, key=lambda x: int(x[2]))[2]:<8}{max(info, key=lambda x: int(x[2]))[2]:<8}")
```

```python
#Task 2.4
from random import randint

def form_squads(roles):
    squads = []
    teamN = -1
    print(roles)
    while all([len(i)>0 for _,i in roles.items()]):
        squads.append([])
        teamN+=1
        for role, info in roles.items():
            i = randint(0, len(info)-1)
            squads[teamN].append(info[i])

            info.pop(i)

    return squads
```

```python
#Task 2.5
squads = form_squads(get_roles(read_file('./Resources/candidates.txt')))

for i in range(len(squads)):
    squad = squads[i]
    print(f"Squad {i+1}, Avg.Score ({sum([int(x[2]) for x in squad])/len(squad):.4g})")
    for person in squad:
        print(f"({person[0]},{person[1]})")
    print()
```

```python
#Task 2.6
data = read_file('./Resources/candidates.txt')
squads = form_squads(get_roles(data))


for squad in squads:
    for person in squad:
        for i in range(len(data)):
            info = data[i]
            if person[0] == info[0]:
                data.pop(i)
                break

for person in data:
    print(f"{person[0]}, who is a {person[1]} with a score of {person[2]} is not selected")
```
