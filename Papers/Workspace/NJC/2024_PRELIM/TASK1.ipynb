```python
# Task 1.1
def task1_1(c, n):
    if c == ' ':
        return '!'

    uppercase = [i for i in range(65, 91)]
    lowercase = [i for i in range(97, 123)]

    if n < 1 or n > 26:
        return -1

    if ord(c) in uppercase:
        return chr(uppercase[(uppercase.index(ord(c)) + n) % len(uppercase)])

    if ord(c) in lowercase:
        return chr(lowercase[(lowercase.index(ord(c)) + n) % len(lowercase)])

    return -1

task1_1('x', 10)
```

```python
# Task 1.2
with open('./TASK1DATA.TXT') as f:
    data = f.read()

key = 'Njc'
key_extended = ''.join([key[i % 3] for i in range(len(data))])

encrypted = ''

uppercase = [i for i in range(65, 91)]
lowercase = [i for i in range(97, 123)]

for i in range(len(data)):
    if ord(key_extended[i]) in uppercase:
        diff = ord(key_extended[i]) - uppercase[0]
    elif ord(key_extended[i]) in lowercase:
        diff = ord(key_extended[i]) - lowercase[0]

    diff = diff or 1
    enc = task1_1(data[i], diff)

    if enc != -1:
        encrypted += enc
    else:
        encrypted += data[i]

with open('./ENCRYPTEDMESSAGE.TXT', 'w') as f:
    f.write(encrypted)

# Showing here
with open('./ENCRYPTEDMESSAGE.TXT') as f:
    print(f.read())
```
