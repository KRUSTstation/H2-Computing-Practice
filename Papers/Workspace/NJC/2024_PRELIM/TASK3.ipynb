```python
# Task 3.1
def merge(left, right):
    sorted_list = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            sorted_list.append(left[i])
            i += 1
        else:
            sorted_list.append(right[j])
            j += 1

    sorted_list.extend(left[i:])
    sorted_list.extend(right[j:])

    return sorted_list

def task3_1(list_of_values):
    if len(list_of_values) <= 1:
        return list_of_values

    mid = len(list_of_values) // 2
    left = task3_1(list_of_values[:mid])
    right = task3_1(list_of_values[mid:])

    return merge(left, right)

task3_1([56, 25, 4, 98, 0, 18, 4, 5, 7, 0]) == [0, 0, 4, 4, 5, 7, 18, 25, 56, 98]
```

```python
# Task 3.2
def task3_2(filename_in, filname_out, no_of_sample):
    from random import sample

    with open(filename_in) as f:
        data = [float(i.strip()) for i in f.readlines()]

    data_sample = sample(data, no_of_sample)

    with open(filname_out, 'w') as f:
        f.writelines([str(i) + '\n' for i in data_sample])

    avg = sum(data_sample) / len(data_sample)

    return len([i for i in data if i < avg])
```

```python
# Task 3.3
def task3_2_sort(filename_in, filname_out, no_of_sample):
    with open(filename_in) as f:
        data = [float(i.strip()) for i in f.readlines()]
        data = task3_1(data)

    data_sample = data[:no_of_sample]

    with open(filname_out, 'w') as f:
        f.writelines([str(i) + '\n' for i in data_sample])

    avg = sum(data_sample) / len(data_sample)

    return len([i for i in data if i < avg])

no_of_sample = 700
alpha = 1000

length = 0

while length < alpha:
    length = task3_2('TASK3FILE.TXT', 'NOTLOWER.TXT', no_of_sample)

length = task3_2_sort('TASK3FILE.TXT', 'LOWER.TXT', no_of_sample)
print(length < alpha)
```
