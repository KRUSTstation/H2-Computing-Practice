```python
# Task 2.1
class LinkedList:
    def __init__(self):
        self.first = None

    def get_first(self):
        return self.first
    def set_first(self, first):
        self.first = first

    def insert(self, data):
        if not self.first:
            self.first = [data, None]
            return

        current = self.first
        while current[1]:
            current = current[1]

        current[1] = [data, None]

    def delete(self, data):
        if not self.first:
            return

        if self.first[0] == data:
            item = self.first
            self.first = self.first[1]
            item[1] = None
            return item

        current = self.first[1]
        previous = self.first
        while current:
            if current[0] == data:
                item = current
                previous[1] = current[1]

                current[1] = None
                return current

            previous = current
            current = current[1]

        return

    def search(self, data):
        if not self.first:
            return False

        if self.first[0] == data:
            return True

        current = self.first
        while current:
            if current[0] == data:
                return True

            current = current[1]

        return False

    def count(self):
        if not self.first:
            return 0

        count = 1
        current = self.first
        while current[1]:
            current = current[1]
            count += 1

        return count

    def to_String(self):
        if not self.first:
            return '[]'

        current = self.first

        string = ''
        string+= '['
        while current[1]:
            string+= f'{current[0]}, '
            current = current[1]
        string+= str(current[0])
        string+= ']'

        return string

LL = LinkedList()
with open('./TASK2DATA.TXT') as f:
    data = [i.strip() for i in f.readlines()]
    for i in data:
        LL.insert(int(i))

print(LL.to_String())
```

```python
# Task 2.2
class Polynomial(LinkedList):
    def create(self, arr):
        arr = sorted(arr, key=lambda x: x[1], reverse=True)
        for i in arr:
            self.insert(i)

    def to_String(self):
        if not self.first:
            return '[]'

        current = self.first
        string = ''
        while current[1]:
            string += f'({current[0][0]})*x**{current[0][1]} + '
            current = current[1]
        string += f'({current[0][0]})*x**{current[0][1]}'

        return string

P = Polynomial()
P.create([(1, 2), (1, 3), (-5, 0)])

Q = Polynomial()
Q.create([(1, 0), (-1, 4), (2, 1)])

print(P.to_String())
print(Q.to_String())
```

```python
# Task 2.3
class Polynomial(LinkedList):
    def create(self, arr):
        arr = sorted(arr, key=lambda x: x[1], reverse=True)
        for i in arr:
            self.insert(i)

    def to_String(self):
        if not self.first:
            return '[]'

        current = self.first
        string = ''
        while current[1]:
            string += f'({current[0][0]})*x**{current[0][1]} + '
            current = current[1]
        string += f'({current[0][0]})*x**{current[0][1]}'

        return string

    def addPoly(self, pol):
        arr = []

        pol1_pointer = self.first
        pol2_pointer = pol.first
        while pol1_pointer and pol2_pointer:
            if pol1_pointer[0][1] == pol2_pointer[0][1]:
                if pol1_pointer[0][0] + pol2_pointer[0][0] != 0:
                    arr.append((pol1_pointer[0][0] + pol2_pointer[0][0], pol1_pointer[0][1]))
                pol1_pointer = pol1_pointer[1]
                pol2_pointer = pol2_pointer[1]

            elif pol1_pointer[0][1] > pol2_pointer[0][1]:
                arr.append(pol1_pointer[0])
                pol1_pointer = pol1_pointer[1]

            else:
                arr.append(pol2_pointer[0])
                pol2_pointer = pol2_pointer[1]

        while pol1_pointer:
            arr.append(pol1_pointer[0])
            pol1_pointer = pol1_pointer[1]

        while pol2_pointer:
            arr.append(pol2_pointer[0])
            pol2_pointer = pol2_pointer[1]

        P = Polynomial()
        P.create(arr)

        return P

P = Polynomial()
P.create([(1, 2), (1, 3), (-5, 0)])

Q = Polynomial()
Q.create([(1, 0), (-1, 4), (2, 1)])

added = P.addPoly(Q)
print(added.to_String())
```
