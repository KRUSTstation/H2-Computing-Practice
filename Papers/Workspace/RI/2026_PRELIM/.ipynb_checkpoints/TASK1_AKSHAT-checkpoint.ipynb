```python
# Task 1.1
def LoadWarehouse(Filename):
    with open(f'./ResourceFiles/{Filename}') as f:
        data = [i.replace('\n', '') for i in f.readlines()]
        data = [i.split(' ') for i in data]

    for i, row in enumerate(data):
        for j, item in enumerate(row):
            if item == 'R':
                return data, i, j

print(LoadWarehouse('WAREHOUSE1.txt'))
```

```python
# Task 1.2
def MoveRobot(Warehouse, RobotRow, RobotCol, Direction):
    directions = {
        'U': (-1, 0),
        'D': (1, 0),
        'R': (0, 1),
        'L': (0, -1)
    }

    d = directions[Direction]
    targetRow = RobotRow + d[0]
    targetCol = RobotCol + d[1]

    if targetRow < 0 or targetCol < 0:
        print("Invalid move")
        return RobotRow, RobotCol

    Warehouse[RobotRow][RobotCol] = 'C'
    Warehouse[targetRow][targetCol] = 'R'

    return targetRow, targetCol

def display(wh):
    for i in wh:
        for j in i:
            print(j, end=' ')
        print()
    print()
    
wh, rr, rc = LoadWarehouse('WAREHOUSE1.txt')
display(wh)

rr, rc = MoveRobot(wh, rr, rc, 'U')
display(wh)
rr, rc = MoveRobot(wh, rr, rc, 'D')
display(wh)
rr, rc = MoveRobot(wh, rr, rc, 'L')
display(wh)
rr, rc = MoveRobot(wh, rr, rc, 'R')
display(wh)
```

```python
# Task 1.3
def ScanNearby(Warehouse, RobotRow, RobotCol):
    scans = [(-1, 0), (1, 0), (0, 1), (0, -1)]

    dirty = clean = shelves = boundary = 0
    for direction in scans:
        targetRow = RobotRow + direction[0]
        targetCol = RobotCol + direction[1]

        if targetRow < 0 or targetCol < 0:
            boundary +=1

        else:
            target = Warehouse[targetRow][targetCol]

            if target == '.':
                dirty += 1
            elif target == '#':
                shelves += 1
            elif target == 'C':
                clean += 1

    return dirty, clean, shelves, boundary
```

```python
# Task 1.4
SuccessfulMoves = 0
InvalidMoves = 0

wh, rr, rc = LoadWarehouse('WAREHOUSE1.txt')
display(wh)

cmd = None
while cmd != 'Q':
    cmd = input('Give command: ').upper()

    if cmd in ['U', 'D', 'R', 'L']:
        old_rr, old_rc = rr, rc
        
        rr, rc = MoveRobot(wh, rr, rc, cmd)

        if rr == old_rr and rc == old_rc:
            InvalidMoves += 1
        else:
            SuccessfulMoves += 1

        display(wh)

    elif cmd == 'S':
        print(ScanNearby(wh, rr, rc))

    elif cmd != 'Q':
        print("Invalid command")

print(f'Successful moves: {SuccessfulMoves}')
print(f'Invalid moves: {InvalidMoves}')
print(f'Dirty tiles remaing: {sum([len([j for j in i if j == '#']) for i in wh])}')
```
