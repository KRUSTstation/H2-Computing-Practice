```python
# Task 4.1
class Presentation:
    def __init__(self, order_n, team_n, proj_t, cat):
        self.OrderNumber = order_n
        self.TeamName = team_n
        self.ProjectTitle = proj_t
        self.Category = cat
        
        self.NextPresentation = None

    def getOrderNumber(self):
        return self.OrderNumber

    def getTeamName(self):
        return self.TeamName

    def getProjectTitle(self):
        return self.ProjectTitle

    def getCategory(self):
        return self.Category

    def getNext(self):
        return self.NextPresentation

    def setNext(self, P):
        self.NextPresentation = P

    def display(self):
        print(f"{self.getOrderNumber()}, {self.getTeamName()}, {self.getProjectTitle()}, {self.getCategory()}")

P = Presentation(3, 'Alpha', 'Smart light', 'Sustainability')
P.display()
```

```python
# Task 4.2
class PresentationSchedule:
    def __init__(self):
        self.FirstPresentation = None

    def getFirst(self):
        return self.FirstPresentation
    def setFirst(self, P):
        self.FirstPresentation = P

    def AddPresentation(self, NewPresentation):
        self.FirstPresentation = self._AddPresentation(NewPresentation, self.FirstPresentation)
        
    def _AddPresentation(self, NewPresentation, first):
        if not first:
            return NewPresentation

        if NewPresentation.getOrderNumber() > first.getOrderNumber():
            first.setNext(self._AddPresentation(NewPresentation, first.getNext()))

        elif NewPresentation.getOrderNumber() < first.getOrderNumber():
            NewPresentation.setNext(first)
            return NewPresentation

        else:
            print('Duplicate OrderNumber')

        return first

    def Summary(self):
        current = self.FirstPresentation
        count = 0
        presentations = {}

        while current:
            current.display()

            if presentations.get(current.getCategory()) is None:
                presentations[current.getCategory()] = 1
            else:
                presentations[current.getCategory()] += 1
            count += 1

            current = current.getNext()

        print()
        print(count)
        print()
        for k,v in presentations.items():
            print(f"{k}: {v}")
```

```python
# Task 4.3
sch = PresentationSchedule()

sch.AddPresentation(Presentation(3, 'Alpha', 'Smart light', 'Sustainability'))
sch.AddPresentation(Presentation(1, 'Beta', 'AI maintenance', 'Digital Tech'))
sch.AddPresentation(Presentation(5, 'NexRetail', 'AR shopping', 'Digital Tech'))
sch.AddPresentation(Presentation(2, 'VitalScan', 'ChroniCare', 'Digital Tech'))
sch.AddPresentation(Presentation(4, 'Microtrust', 'Credit scoring for SMEs', 'FinTech'))
sch.AddPresentation(Presentation(2, 'EnviroGrid', 'Carbon tracing for maritime', 'Sustainability'))

sch.Summary()
```
