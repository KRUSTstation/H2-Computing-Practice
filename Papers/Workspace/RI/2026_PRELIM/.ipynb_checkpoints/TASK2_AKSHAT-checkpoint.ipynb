```python
# Task 2.1
import csv

def merge(left, right):
    sorted_arr = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i][0] < right[j][0]:
            sorted_arr.append(left[i])
            i+=1
        elif left[i][0] > right[j][0]:
            sorted_arr.append(right[j])
            j+=1
            
        else:
            if left[i][2] <= right[j][2]:
                sorted_arr.append(left[i])
                i+=1
            
            else:
                sorted_arr.append(right[j])
                j+=1

    sorted_arr.extend(left[i:])
    sorted_arr.extend(right[j:])
    return sorted_arr

def MergeSort(Registrations, First, Last):
    if First >= Last:
        return Registrations[First:Last+1]

    mid = (First + Last) // 2
    left = MergeSort(Registrations, First, mid)
    right = MergeSort(Registrations, mid+1, Last)

    return merge(left, right)

with open('./ResourceFiles/REGISTRATIONS.txt') as f:
    data = list(csv.reader(f))

print(data[:10])
print()
print(MergeSort(data, 0, len(data) - 1)[:10])
```

```python
# Task 2.2
def BinarySearch(Registrations, CourseCode):
    low = 0 
    high = len(Registrations) - 1

    while low <= high:
        mid = (low + high) // 2

        if CourseCode < Registrations[mid][0]:
            high = mid - 1
        elif CourseCode > Registrations[mid][0]:
            low = mid + 1

        else:
            return mid

    return -1

found = BinarySearch(MergeSort(data, 0, len(data) - 1),"CS101")
if found == -1:
    print('Course not found')
else:
    print(found)

found = BinarySearch(MergeSort(data, 0, len(data) - 1),"CS999")
if found == -1:
    print('Course not found')
else:
    print(found)
```

```python
# Task 2.3
def DisplayCourseList(Registrations, CourseCode):
    students = []
    
    found = BinarySearch(Registrations, CourseCode)
    while found != -1:
        students.append(Registrations[found][2])
        
        Registrations.pop(found)
        found = BinarySearch(Registrations, CourseCode)

    if len(students) <= 0:
        print(f'{CourseCode}: Course not found')

    return students

print(DisplayCourseList(MergeSort(data, 0, len(data) - 1), "CS302"))
print(DisplayCourseList(MergeSort(data, 0, len(data) - 1), "CS999"))
```

```python
# Task 2.4
def CourseStatistics(Registrations):
    courses = {}
    
    for item in Registrations:
        if item[0] not in courses.keys():
            courses[item[0]] = 0
        else:
            courses[item[0]] += 1

    largest = None
    largest_students = None
    for course, students in courses.items():
        print(f'{course}: {students} students')
        
        if not largest or students > largest_students[0]:
            largest = [course]
            largest_students = [students]

        elif students == largest_students[0]:
            largest.append(course)

    print()
    print(f'Total courses: {len(courses.keys())}')
    print()

    print(f'Largest classes:\n{largest}')

CourseStatistics(MergeSort(data, 0, len(data) - 1))
```
