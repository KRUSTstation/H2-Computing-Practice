```python
# Task 3.1

class Board:
    
    def __init__(self):
        self.data = []
        for i in range(1,9): # numbers 1 - 8
            self.data.append(i)
        self.data.append(0) # 0 for blank tile
        
        self.blank = (2,2)
        
        
    def __str__(self):
        '''
            For debugging
        '''
        return f"{self.data}, Blank: {self.blank}"
    
    def get(self, x, y):
        i = 3*y + x
        return self.data[i]
    
    def set(self, x, y, n):
        i = 3*y + x
        self.data[i] = n
        
```

```python
# Task 3.2
class SlidingGame(Board):
    
    def is_valid(self, x, y):
        if 0 <= x <= 2 and 0 <= y <= 2:
            horizontally_adjacent = (x == self.blank[0] + 1 or x == self.blank[0] - 1) and y == self.blank[1]
            vertically_adjacent = (y == self.blank[1] + 1 or y == self.blank[1] - 1) and x == self.blank[0]
            if horizontally_adjacent or vertically_adjacent:
                return True
            else:
                # print(f"({x}, {y}) is not adjacent to blank {self.blank}")
                return False
        else:
            # print(f"({x}, {y}) is not a valid coordinate")
            return False
        
    def slide(self, x, y):
        if self.is_valid(x, y):
            num = self.get(x, y)
            self.set(self.blank[0], self.blank[1], num)
            self.set(x, y, 0)
            self.blank = (x, y)
            return True
        else:
            return False
        
    def options(self):
        result = []
        for idx, num in enumerate(self.data):
            y = idx // 3
            x = idx % 3
            if self.is_valid(x, y):
                result.append((x,y))
        
        return result
    
    def is_complete(self):
        if self.data[8] == 0 and self.data[:8] == sorted(self.data[:8]):
            return True
        else:
            return False
        
    def display(self):
        for y in range(3):
            for x in range(3):
                print(self.data[3*y+x], end= " ")
            print()
            print("-"*6)
```

```python
s = SlidingGame()
s.display()
print(s.is_complete())
s.slide(2,1)
print("After sliding (2,1)")
s.display()
s.slide(1,1)
print("After sliding (1,1)")
s.display()
print(s.is_complete())
```

```python
# Task 3.3

def shuffle_board(board):
    import random 
    
    while board.data[0] == 1: # repeat until the first array value is no longer 1
        options = board.options()  # check valid options
        x, y = random.choice(options)
        board.slide(x, y)  # pick random move

s = SlidingGame()
print("Before shuffling:")
s.display()
shuffle_board(s)

print("After shuffling:")
s.display()
```

```python
# Task 3.4
moves = 0

while not s.is_complete():
    # Display the board
    s.display()
    
    # prompt player to pick valid tiles to slide
    options = s.options()

    while True:
        try:
            print("Here are the valid tiles to slide:")
            print("Option number | Tile number")
            for idx, option in enumerate(options):
                print(f"{idx:<14}: {s.get(*option)}")
            user_input = input("Please key in the option number: ")

            # validate user input
            if  0<= int(user_input) <= len(options):
                x, y = options[int(user_input)]
                
                # slide the tile
                s.slide(x, y)
                moves += 1
                break
        except:
            print("Invalid input")

print(f"Congrats! It took you {moves} moves.")
    
```
