```python
# Task 1.1

def task1_1(artefacts_list):
    
    valid_artefacts = []

    for i in range(len(artefacts_list)):
        artefact_id = artefacts_list[i][0]
        rating = artefacts_list[i][1]

        try:
            if not(artefact_id[0].isupper() and 1000 <= int(artefact_id[1:]) <= 9999):
                raise ValueError("ID does not start with uppercase letter and in the range between 1000 and 9999.")

            if not 1 <= int(rating) <= 100:
                raise ValueError("Rating must be an integer between 1 and 100.")
                
            valid_artefacts.append((artefact_id, int(rating)))

        except ValueError as error:
            print(f"Invalid artefact: Artefact ID: {artefact_id} Error: {error}")

    return valid_artefacts
```

```python
artefacts_list = [("A1234", 85),("b1234", 85),("C1234", 85),("B0567", 92),("A3891", 145),("C4012", 78),("B5678", 95),("D6543", 88),("E7890", 65),("A8123", 73),("C9012", 98),("B1567", 52)]
valid_artefacts = task1_1(artefacts_list)

print(f"Total number of valid artefacts: {len(valid_artefacts)}")

print("\nFirst 5 artefacts:")
for i in range(5):
    print(valid_artefacts[i])
```

```python
# Task 1.2

def quicksort(array):
    
    if len(array) <= 1:
        return array

    less_than = []
    more_than = []
    equal = []  

    pivot = array[0][1] # Sort by rating

    for item in array: # item is a tuple (id, rating)
        if item[1] < pivot:
            less_than.append(item)
        elif item[1] > pivot:
            more_than.append(item)
        else:
            equal.append(item)  
            
    # Sort in descending order
    return quicksort(more_than) + equal + quicksort(less_than)

def task1_2(artefacts_list):

    if not artefacts_list: # if the list is empty
        return []
    
    return quicksort(artefacts_list)
```

```python
sorted_result = task1_2(valid_artefacts)

print("Sorted artefacts (by rating in descending order):")
for i in sorted_result:
    print(i)
```

```python
# Task 1.3

def binary_search_desc(array, target, low, high):

    if low > high:
        return None

    mid = (low + high) // 2
    current_rating = array[mid][1]
    
    results = []

    if current_rating == target:
        results.append(array[mid])

        # There may be more than 1 artefact with the same rating
        # look to left
        left = mid - 1
        while left >= 0 and array[left][1] == target:
            results.append(array[left])
            left -= 1

        # look to right
        right = mid + 1
        while right < len(array) and array[right][1] == target:
            results.append(array[right])
            right += 1
        return results

    elif current_rating < target: # target is on the left (descending)
        return binary_search_desc(array, target, low, mid - 1)
    else:
        return binary_search_desc(array, target, mid + 1, high)


def task1_3(artefacts_list, target_rating):

    sorted_artefacts = task1_2(artefacts_list)  

    results = binary_search_desc(sorted_artefacts, target_rating, 0, len(sorted_artefacts) - 1)
    return results
```

```python
search_values = [85, 100]

for target_rating in search_values:
    print(f"Searching for artefact rating {target_rating}:")
    results = task1_3(valid_artefacts, target_rating)

    if results:
        print(f"Found {len(results)} artefact(s) with rating {target_rating}:")
        for artefact in results:
            print(f"ID: {artefact[0]}, Rating: {artefact[1]}")
    else:
        print(f"No artefacts found with rating {target_rating}")
    
    print()
```

```python
# Task 1.4

def task1_4(artefacts_list):

    if not artefacts_list: # empty list
        return None
    
    total_rating = sum(artefact[1] for artefact in artefacts_list)
    average = round(total_rating/len(artefacts_list), 2)

    above_average = sum(1 for artefact in artefacts_list if artefact[1] > average)
    
    sorted_artefacts = sorted(artefacts_list, key=lambda x: x[1])
    
    least_3 = [artefact[0] for artefact in sorted_artefacts[:3]]
    
    return (average, above_average, least_3)
```

```python
result = task1_4(valid_artefacts)

if result:
    average, above_count, least_3 = result
    print(f"Average historical significance: {average}")
    print(f"Number of artefacts above average: {above_count}")
    print(f"The 3 least significant artefacts: {', '.join(least_3)}")
else:
    print("No artefacts to analyse")
```
