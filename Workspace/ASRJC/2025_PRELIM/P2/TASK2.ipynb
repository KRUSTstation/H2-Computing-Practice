```python
#Task 2.1
def generate_runners():
    from random import choice
    from random import randint

    runners = []
    for _ in range(15000):
        n = randint(80000000, 98999999)
        while n in runners:
            n = randint(80000000, 98999999)
        
        runners.append((n,choice(["Men's Marathon", "Women's Marathon", "Men's Half-Marathon", "Women's Half-Marathon"])))

    with open('./runners.txt', 'w') as f:
        f.writelines([f"{i[0]},{i[1]}" for i in runners])

    return runners

generate_runners()
```
