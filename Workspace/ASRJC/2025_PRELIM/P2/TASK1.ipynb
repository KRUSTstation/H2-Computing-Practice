```python
# Task 1.1
def encode_str(text, token):
    new_str = ""
    prev_char = ""
    consecutive = 0

    for char in text:
        consecutive+=1
        if char != prev_char:
            if consecutive >= 3:
                new_str+=f"{token}{prev_char}{consecutive}"
            else:
                new_str+=f"{prev_char * consecutive}"

            consecutive = 0

        prev_char = char

    consecutive+=1
    if consecutive >= 3:
        new_str+=f"{token}{prev_char}{consecutive}"
    else:
        new_str+=f"{prev_char * consecutive}"

    return new_str

encode_str('aaaaabbcccddd', '%')
```

```python
#Task 1.2
def decode_str(encoded, token):
    decoded = ""
    i = 0
    L = len(encoded)
    while i < L:
        if encoded[i] == token and i + 2 < L and encoded[i+2].isdigit():
            char = encoded[i+1]
            count = int(encoded[i+2])
            decoded += char * count
            i += 3
        else:
            decoded += encoded[i]
            i += 1
    return decoded

decode_str('%a5bb%c3%d3', '%')
```

```python
#Task 1.3
with open('./Resource Files/encoding.txt') as f:
    data = [i.strip() for i in f.readlines()]

decoded = [decode_str(i, '@') for i in data]
for i in decoded:
    print(i)
```
