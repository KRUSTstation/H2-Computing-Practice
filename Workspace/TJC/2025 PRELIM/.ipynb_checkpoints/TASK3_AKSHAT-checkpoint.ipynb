```python
# Task 3.1
class Board:
    def __init__(self):
        self.data = [[1,2,3],
                     [4,5,6],
                     [7,8,0]]
        self.blank = (2, 2)

    def get(self, x, y):
        return self.data[y][x]

    def set(self, x, y, n):
        self.data[y][x] = n
```

```python
# Task 3.2
class SlidingGame(Board):
    def __init__(self):
        super().__init__()

    def is_valid(self, x, y):
        bx, by = self.blank

        if not (0 <= x < len(self.data[0]) and 0 <= y < len(self.data)):
            return False
    
        return abs(x - bx) + abs(y - by) == 1

    def slide(self, x, y):
        if not self.is_valid(x, y):
            return False

        self.set(self.blank[0], self.blank[1], self.get(x, y))
        self.set(x, y, 0)
        self.blank = (x, y)

    def options(self):
        valid = []

        for y, arr in enumerate(self.data):
            for x, val in enumerate(arr):
                if self.is_valid(x, y):
                    valid.append((x, y))
        
        return valid

    def is_complete(self):
        if self.blank[1] == len(self.data)-1 and self.blank[0] == len(self.data[0])-1:
            prev = 0
            for y, arr in enumerate(self.data):
                for x, val in enumerate(arr):
                    if self.blank[1] == y and self.blank[0] == x:
                        continue
                    
                    if not self.get(x, y) == prev+1:
                        return False
                    prev = self.get(x, y)
            return True
            
        return False

g = SlidingGame()
print(g.get(1,2))
print(g.is_complete())
g.slide(1,2)
```

```python
# Task 3.3
import random

def shuffle_board(board: SlidingGame):
    while board.get(0,0) == 1:
        c = random.choice(board.options())
        board.slide(c[0], c[1])

s = SlidingGame()
print(s.data)
shuffle_board(s)
print(s.data)
```

```python
# Task 3.4
def display(board):
    for y in board.data:
        for x in y:
            print(x, end=' ')
        print()

moves = 0
while not s.is_complete():
    display(s)
    print('Valid: ')
    print(s.options())
    x, y = [int(i) for i in input('Pick a valid tile: ').split(',')]
    while not s.is_valid(x, y):
        x, y = [int(i) for i in input('Invalid! Pick a valid tile: ').split(',')]

    s.slide(x, y)
    moves+=1

print(f"Used {moves} moves")
```
