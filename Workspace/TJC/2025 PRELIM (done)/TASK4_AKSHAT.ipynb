```python
# Task 4.1
import sqlite3
import csv

pref = './P2 Resource files/'

conn = sqlite3.connect(f'{pref}Busroutes.db')
cursor = conn.cursor()

with open(f'{pref}Route.csv') as f:
    data = [i for i in csv.reader(f)][1:]

    for i in data:
        cursor.execute('INSERT INTO Route(ServiceNo, Direction, StopSequence, BusStopCode) VALUES (?, ?, ?, ?)', i)

with open(f'{pref}Service.csv') as f:
    data = [i for i in csv.reader(f)][1:]

    for i in data:
        cursor.execute('INSERT INTO Service(ServiceNo, Operator, Category) VALUES (?, ?, ?)', i)

with open(f'{pref}Stop.csv') as f:
    data = [i for i in csv.reader(f)][1:]

    for i in data:
        cursor.execute('INSERT INTO Stop(RoadName, Description, BusStopCode) VALUES (?, ?, ?)', (i[1], i[2], i[0]))

conn.commit()
conn.close()
```

```python
# Task 4.2
import sqlite3

bus_no = input("Gib bus number: ")

conn = sqlite3.connect(f'{pref}Busroutes.db')
cursor = conn.cursor()

cursor.execute('SELECT r.Direction, r.BusStopCode, st.RoadName, st.Description FROM Route AS r JOIN Stop AS st ON r.BusStopCode = st.BusStopCode WHERE r.ServiceNo = ? ORDER BY r.Direction AND r.StopSequence ASC', (bus_no,))
data = cursor.fetchall()
print(data)

conn.close()
```

```python
# Task 4.3
from flask import Flask, render_template, request
import sqlite3

app = Flask(__name__, template_folder='TASK_4_3')

pref = './P2 Resource files/'

@app.route('/', methods=['GET', 'POST'])
def base():
    if request.method == 'POST':
        bus_stop = request.form.get('bus_stop')
        conn = sqlite3.connect(f'{pref}Busroutes.db')
        cursor = conn.cursor()

        cursor.execute('SELECT r.ServiceNo, s.Operator FROM Route AS r JOIN Service as s ON r.ServiceNo = s.ServiceNo WHERE r.BusStopCode = ?', (bus_stop,))
        data = cursor.fetchall()
        
        conn.close()
        if not data: data = 'Invalid bus stop'

        return render_template('base.html', data=data)
        
    return render_template('base.html')

app.run()
```
