```python
# Task 1.1
def task_1_1():
    import csv
    
    with open('IDNUMBER_NAME1.CSV') as f:
        data = [i for i in csv.reader(f)]

    Max = 20
    customer_array = [None]*(Max)
    for person in data:
        index = int(person[0]) % Max
        customer_array[index] = tuple(person)
    for i, item in enumerate(customer_array):
        print(f"{i:5} | {item}")

task_1_1()
```

```python
# Task 1.2
def task_1_2():
    import csv
    
    with open('IDNUMBER_NAME2.CSV') as f:
        data = [i for i in csv.reader(f)]

    Max = 20
    customer_array = [None]*(Max)
    for i, person in enumerate(data):
        index = int(person[0]) % Max

        while customer_array[index]:
            index = index+1
        customer_array[index] = tuple(person)
        
    for i, item in enumerate(customer_array):
        print(f"{i:5} | {item}")

task_1_2()
```

```python
# Task 1.3
def task_1_3():
    import csv
    
    with open('IDNUMBER_NAME2.CSV') as f:
        data = [i for i in csv.reader(f)]

    Max = 20
    customer_array = [None]*(Max)
    for i, person in enumerate(data):
        index = int(person[0]) % Max

        while customer_array[index]:
            index = index+1
        customer_array[index] = tuple(person)
        
    user_input = input("Input an ID number: ")
    for i, person in enumerate(customer_array):
        if person and person[0] == user_input:
            print(f'Found in index pos: {i}')
            return
    print('Not Found')

task_1_3()
```
