```python
# Task 3.1
def val_hexa(h):
    digits = h[1:]
    if h[0] != '#':
        return False
    elif not all(i in '1234567890ABCDEFabcdef' for i in digits):
        return False
    else:
        return True

while not val_hexa(hexa:=input('input valid hexadecimal: ')):
    print('Invalid hexa')

```
