```python
# Task 2.2
class Book:
    def __init__(self, ID, t, p):
        self.id = ID
        self.title = t
        self.price = p

    def set_id(self, ID):
        self.id = ID

    def set_title(self, s):
        self.title = s

    def set_price(self, p):
        self.price = p

    def get_id(self):
        return self.id

    def get_title(self):
        return self.title

    def get_price(self):
        return self.price

    def summary(self):
        return f'{self.id}, {self.title}, {self.price}'

    def compare_with(self, bk, option):
        if option == 'p':
            if self.get_price() < bk.get_price():
                return -1
            elif self.get_price() == bk.get_price():
                return 0
            else:
                return 1
        if option == 'a':
            if self.get_title().lower() < bk.get_title().lower():
                return -1
            elif self.get_title().lower() == bk.get_title().lower():
                return 0
            else:
                return 1
        
book_1 = Book(1617294632, 'Natural Language Processing in Action: Understanding analyzing and generating text with Python', 64.42)
book_2 = Book(9781617294433, 'Deep Learning with Python', 30.87)
book_3 = Book(1492032646, 'Hands-on Machine Learning with Scikit-Learn Keras and TensorFlow: Concepts Tools and Techniques to Build Intelligent Systems', 49.73)
book_4 = Book(1616558458, 'World of Warcraft: Chronicle Volume 1', 58.34)

books = []
while (preference := input('What is your ordering preference? (t or p): ').lower()) not in ('t', 'p'):
    print('Invalid input')

if preference == 't':
    preference = 'a'

inventory = [book_1, book_2, book_3, book_4]
for book in inventory:
    if not books:
        books.append(book)
    else:
        inserted = False
        for i, other_book in enumerate(books):
            if other_book.compare_with(book, preference) == 1:
                books.insert(i, book)
                inserted = True
                break
        if not inserted:
            books.append(book)

for book in books:
    print(book.summary())
```

```python
# Task 2.3
class Book:
    def __init__(self, ID, t, p):
        self.id = ID
        self.title = t
        self.price = p

    def set_id(self, ID):
        self.id = ID

    def set_title(self, s):
        self.title = s

    def set_price(self, p):
        self.price = p

    def get_id(self):
        return self.id

    def get_title(self):
        return self.title

    def get_price(self):
        return self.price

    def summary(self):
        return f'{self.id}, {self.title}, {self.price}'

    def compare_with(self, bk, option):
        if option == 'p':
            if self.get_price() < bk.get_price():
                return -1
            elif self.get_price() == bk.get_price():
                return 0
            else:
                return 1
        if option == 'a':
            if self.get_title().lower() < bk.get_title().lower():
                return -1
            elif self.get_title().lower() == bk.get_title().lower():
                return 0
            else:
                return 1

class ElectronicBook(Book):
    def __init__(self, ID, t, p, pg):
        super().__init__(ID, t, p)
        self.pages = pg

    def set_pages(self, pg):
        self.pages = pg

    def get_pages(self):
        return self.pages

    def summary(self):
        return f'{super().summary()}, {self.pages} pages'

    def compare_with(self, bk, option):
        if isinstance(bk, AudioBook):
            return -1
        else:
            return super().compare_with(bk, option)
            
class AudioBook(Book):
    def __init__(self, ID, t, p, d):
        super().__init__(ID, t, p)
        self.duration = d

    def set_duration(self, d):
        self.duration = d

    def get_duration(self):
        return self.duration

    def summary(self):
        return f'{super().summary()}, {self.duration} hours'

    def compare_with(self, bk, option):
        if isinstance(bk, ElectronicBook):
            return 1
        else:
            return super().compare_with(bk, option)

import csv

with open('TASK2_3.TXT') as f:
    data = [[item.replace(' ', '') for item in i] for i in csv.reader(f)]
    inventory = []
    for bookInfo in data:
        if bookInfo[3] == 'EBOOK':
            inventory.append(ElectronicBook(bookInfo[0], bookInfo[1], float(bookInfo[2]), int(bookInfo[4].replace('pages', ''))))
        elif bookInfo[3] == 'AUDIOBOOK':
            inventory.append(AudioBook(bookInfo[0], bookInfo[1], float(bookInfo[2]), int(bookInfo[4].replace('hours', ''))))

books = []
while (preference := input('What is your ordering preference? (t or p): ').lower()) not in ('t', 'p'):
    print('Invalid input')

if preference == 't':
    preference = 'a'

for book in inventory:
    if not books:
        books.append(book)
    else:
        inserted = False
        for i, other_book in enumerate(books):
            if other_book.compare_with(book, preference) == 1:
                books.insert(i, book)
                inserted = True
                break
        if not inserted:
            books.append(book)

for book in books:
    print(book.summary())
```
