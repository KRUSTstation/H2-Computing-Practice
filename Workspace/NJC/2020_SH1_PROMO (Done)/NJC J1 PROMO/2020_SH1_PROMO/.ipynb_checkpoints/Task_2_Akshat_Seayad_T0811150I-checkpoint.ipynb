```python
#Task 2.1
a = int(input('input a'))
r = int(input('input r'))
m = int(input('input m'))

counter = 0
for n in range(1, m+1):
    print(a*r**n, end = ' ')
    counter+=a*r**n

print('\n'+str(counter))
```

```python
# Task 2.2
def Fibonacci(n):
    if n<0:
        print("Incorrect input")
        return
    elif n==0:
        return 0
    elif n==1:
        return 1
    a, b = 0, 1
    for _ in range(2, n):
        a, b = b, a+b
    return b
    
Fibonacci(10)
```
