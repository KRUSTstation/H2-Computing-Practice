```python
#Task 1.1
import csv

def read_data(filename):
    with open(f'./RENAME_FOLDER/{filename}') as f:
        data = [i for i in csv.reader(f, delimiter = ';')]
    return data
    
print(read_data('PARTICIPANTS.TXT'))
```

```python
# Task 1.2
def gender_count(participants):
    males = len([person for person in participants if person[2]=='M'])
    females = len([person for person in participants if person[2]=='F'])
    return males, females

print(gender_count(read_data('PARTICIPANTS.TXT')))
```

```python
# Task 1.3
def role_statistics(participants):
    from collections import Counter

    data = dict(Counter(i[1] for i in participants))
    print(f"{'Role':<20} {'Number'}")
    for role, count in data.items():
        print(f"{role:<20} {count}")
        

role_statistics(read_data('PARTICIPANTS.TXT'))
```

```python
# Task 1.4
def form_group(participants):
    from random import randint

    sorted_dict = {}
    for info in participants:
        if info[1] not in sorted_dict.keys():
            sorted_dict[info[1]] = [info]
        else:
            sorted_dict[info[1]].append(info)
    data = {role: people for role, people in sorted_dict.items() if role in ['Coder', 'Maker', 'Tester', 'Manager', 'Designer']}
    if len(data) < 5:
        return []
    group = [people[randint(0, len(people)-1)] for people in data.values()]
    participants = [i for i in participants if i not in group]
    return group, participants
    
participants_data = read_data('PARTICIPANTS.TXT')
print(form_group(participants_data)[0])
```

```python
print(form_group(participants_data))
```

```python
# Task 1.5
participants = read_data('PARTICIPANTS.TXT')
counter = 1
with open('./RENAME_FOLDER/GROUPS.TXT', 'w') as f:
    while (info:=form_group(participants)) != []:
        group, participants = info
        f.write(f'Group {counter}\n')
        for person in group:
            for info in person:
                f.write(f'{info},')
            f.write('\n')
        counter+=1
```
