```python
#Task 3.1
class Patient:
    def __init__(self, name, NRIC):
        self.name = name
        self.NRIC = NRIC

    def getName(self):
        return self.name

    def getNRIC(self):
        return self.NRIC

    def __str__(self):
        return self.getName()

class Visitor:
    def __init__(self, name, contact_no):
        self.name = name
        self.contact_no = contact_no

    def getName(self):
        return self.name

    def getContact(self):
        return self.contact_no

    def __str__(self):
        return f'{self.getName()}:{self.getContact()}'
```

```python
#Task 3.3
class Queue:
    def __init__(self):
        self.buffer = []

    def enqueue(self, Visitor):
        if Visitor in self.buffer or len(self.buffer) >= 10: # check if person is already in the queue and queue size
            return False
        else:
            self.buffer.append(Visitor)
            return True

    def dequeue(self):
        if self.buffer == []:
            return None
        visitor = self.buffer.pop(0)
        return str(visitor)

    def __str__(self):
        return str([str(item) for item in self.buffer])
```

```python
#Task 3.4
q = Queue()
q.enqueue(Abbott)
q.enqueue(Norby)
print(q)
q.dequeue()
print(q)
Alvinia = Visitor('Alvinia', '955-561-84')
q.enqueue(Alvinia)
print(q)
```

```python
#Task 3.5
class Bed:
    def __init__(self, floor, ward, bedNo, startVisitHr, endVisitHr):
        self.floor = floor
        self.ward = ward
        self.bedNo = bedNo
        self.visitHourStart = startVisitHr
        self.visitHourEnd = endVisitHr
        self.queue = Queue()
        self.occupiedBy = None

    def __str__(self):
        return f'{self.floor}-{self.ward}-{self.bedNo}:{self.occupiedBy.getName()}'
```

```python
#Task 3.6
from datetime import datetime

class Hospital:
    def __init__(self, noFloors, noWards, noBeds):
        bedList = []
        for floor in range(1,noFloors+1):
            for ward in range(1,noWards+1):
                for bed in range(1,noBeds+1):
                    if floor == noFloors:
                        startVisitHr = 17
                        endVisitHr = 19
                    else:
                        startVisitHr = 12
                        endVisitHr = 20
                        
                    bedList.append(Bed(floor, ward, bed, startVisitHr, endVisitHr))
        self.beds = bedList

    def visit(self, visitor, floor, ward, bed, DateTime):
        for patient_bed in self.beds:
            if patient_bed.floor == floor and patient_bed.ward == ward and patient_bed.bedNo == bed:
                if not patient_bed.occupiedBy:
                    return False
                elif patient_bed.visitHourStart <= DateTime.hour <= patient_bed.visitHourEnd:
                    return patient_bed.queue.enqueue(visitor)
                else:
                    return False

    def hash(self, floor, ward, bed):
        for i in range(len(self.beds)):
            if self.beds[i].floor == floor and self.beds[i].ward == ward and self.beds[i].bedNo == bed:
                return i

    def occupy(self, patient, floor, ward, bed):
        i = self.hash(floor, ward, bed)
        self.beds[i].occupiedBy = patient

    def showOccupancy(self):
        for i in range(len(self.beds)):
            if self.beds[i].occupiedBy:
                print(f'{i}->{str(self.beds[i])} {str(self.beds[i].queue)}')
```
