```python
# Task 3.1
class ContainerNode:
    def __init__(self, id, category, weight):
        self.__container_number = id
        self.__category_of_goods = category
        self.__cargo_weight = weight
        self.__next = None

    def get_container_number(self):
        return self.__container_number
    
    def set_container_number(self, data):
        self.__container_number = data

    def get_category_of_goods(self):
        return self.__category_of_goods
    
    def set_category_of_goods(self, data):
        self.__category_of_goods = data

    def get_cargo_weight(self):
        return self.__cargo_weight
    
    def set_cargo_weight(self, data):
        self.__cargo_weight = data

    def get_next(self):
        return self.__next
    
    def set_next(self, data):
        self.__next = data

class LinkedList:
    def __init__(self):
        self.__head = None

    def get_head(self):
        return self.__head
    
    def set_head(self, data):
        self.__head = data

    def is_empty(self):
        return self.get_head() == None

    def insert_to_end(self, container_number, category_of_goods, cargo_weight):
        n = ContainerNode(container_number, category_of_goods, cargo_weight)
        if self.is_empty():
            self.set_head(n)
            return
        
        current = self.get_head()

        while current and current.get_next():
            current = current.get_next()

        current.set_next(n)

    def insert_to_front(self, container_number, category_of_goods, cargo_weight):
        n = ContainerNode(container_number, category_of_goods, cargo_weight)
        old_head = self.get_head()

        self.first = n
        self.first.set_next(old_head)

    def delete_from_end(self):
        if self.is_empty():
            return 'Cannot delete - Empty'
        
        current = self.get_head()
        previous = None

        while current.get_next():
            previous = current
            current = current.get_next()

        previous.set_next(None)

    def delete_from_front(self):
        if self.is_empty():
            return 'Cannot delete - Empty'
        
        next = self.first.get_next()
        self.first.set_next(None)
        self.first = next
```

```python
# Task 3.2
class Stack(LinkedList):
    def __init__(self):
        super().__init__()

    def is_empty(self):
        return self.get_head() == None
    
    def peek(self, container_number):
        if self.is_empty():
            return 'No containers'
        
        current = self.get_head()

        while current:
            if current.get_container_number() == container_number:
                return current
            current = current.get_next()

    def display(self):
        if self.is_empty():
            return 'No containers'
        
        current = self.get_head()

        while current:
            print(current.get_container_number())
            current = current.get_next()
```

```python
# Task 3.3
import csv

with open('./Resources/container_data.csv') as f:
    data = [i for i in csv.reader(f)]

mainStack = Stack()
spareStack = Stack()

for info in data:
    if info[0] == 'INBOUND':
        mainStack.insert_to_end(info[1], info[2], info[3])

mainStack.display()

for info in data:
    if info[0] == 'OUTBOUND':
        if mainStack.peek(info[1]) == 'No containers':
            spareStack.insert_to_end(info[1], info[2], info[3])
        else:
            break
```
