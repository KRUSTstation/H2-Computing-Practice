A denary number can be converted to binary using the division by 2 method. The denary number is divided by 2 repeatedly until it becomes 0. After each division, the remainder of 1 or 0 is stored. These remainders are written in reverse order to create the binary number.

For example, converting the denary number 28 to binary:

```text
28/2=14 remainder 0
14/2=7 remainder 0
7/2=3 remainder 1
3/2=1 remainder 1
1/2=0 remainder 1
28 denary is 11100 in binary.
```

A denary number can be converted to binary using the division by 2 method and a stack.

The stack can store any number of integer data items using a 1-dimensional list. The top of stack pointer stores the index of the next space to store data in the stack.

The data is **not** deleted when it is popped from the stack; the top of stack pointer is changed to point to the next space to store data.

The program must only use local variables; any data required by a function needs to be passed as parameters and returned (where appropriate).

## Task 1.1
The function `push()` takes the stack, the top of stack pointer and the data item to push onto the stack as parameters. The function stores the data item in the stack and updates the top of stack pointer. The function returns the top of stack pointer. If the stack is full, the length of the list should be increased. 

Write program code for the function `push()`.<div style="text-align: right">[4]</div> 

````python
# Task 1.1
def push(stack, top, data):
    if top + 1 >= len(stack):
        stack.append(None)

    stack[top] = data
    top+=1

    return top
````

## Task 1.2
The function `pop()` takes the stack and the top of stack pointer as parameters. The data item at the top of the stack is returned along with the updated top of stack pointer. 

The data is **not** deleted when it is popped. The top of stack pointer is changed to point to the new top of stack; this allows data to be overwritten when new data is pushed onto the stack.

If the stack is empty, the function returns `False`.

Write program code for the function `pop()`.<div style="text-align: right">[4]</div> 

````python
# Task 1.2
def pop(stack, top):
    if len(stack) <= 0:
        return False

    top-=1
    data = stack[top]

    return data, top
````

## Task 1.3

The function `main()` declares a 1-dimensional list to represent the stack and initialises the top of stack pointer to 0.

The function takes a positive integer as input from the user or as a parameter. 

The function only outputs to the screen and does not return any data.

If the number input is 0, the program outputs the binary value 0.

If the number input is greater than 0, the input is repeatedly divided by 2 until it reaches 0; each remainder is pushed onto the stack using the function `push()` you created in Task 1.1.
Each digit is then popped from the stack to create the binary number using the function `pop()` you created in Task 1.2. The binary number is then output on one line.

Write program code for `main()` and call this function. <div style="text-align: right">[10]</div> 

Test your program with the following inputs:

```
Test 1: 0
Test 2: 22
Test 3: 46967
```

Show the output for each test. <div style="text-align: right">[2]</div> 

````python
# Task 1.3
def main(integer):
    stack = []
    top = 0

    print("Binary: ", end='')
    if integer == 0:
        print("0", end='')
    elif integer > 0:
        while integer > 0:
            remainder = integer % 2

            top = push(stack, top, remainder)

            integer //= 2

    while top:
        data, top = pop(stack, top)
        print(data, end='')
    print()
    
main(0)
main(22)
main(46967)
````

````python
# Task 2.1
class Room:
    def __init__(self, ne, ee, se, we):
        self.north_exit = ne
        self.east_exit = ee
        self.south_exit = se
        self.west_exit = we
    
    def get_ne(self):
        return self.north_exit
    def set_ne(self, data):
        self.north_exit = data

    def get_se(self):
        return self.south_exit
    def set_se(self, data):
        self.south_exit = data

    def get_ee(self):
        return self.east_exit
    def set_ee(self, data):
        self.east_exit = data

    def get_we(self):
        return self.west_exit
    def set_we(self, data):
        self.west_exit = data

    def get_exit(self, char):
        if char == 'N':
            return self.north_exit

        if char == 'S':
            return self.south_exit
        
        if char == 'E':
            return self.east_exit
        
        if char == 'W':
            return self.west_exit
        
        return -1
````

````python
class Start(Room):
    def __init__(self, ne, ee, se, we, start_text):
        super().__init__(ne, ee, se, we)

        self.start_text = start_text
    
    def get_start_text(self):
        return self.start_text
    def set_start_text(self, data):
        self.start_text = data

    def get_message(self):
        to_return = ["Start", self.start_text + ' ']

        if self.get_exit('N') != -1:
            to_return[1]+='There is a door to the North. '
        
        if self.get_exit('S') != -1:
            to_return[1]+='There is a door to the South. '
        
        if self.get_exit('E') != -1:
            to_return[1]+='There is a door to the East. '

        if self.get_exit('W') != -1:
            to_return[1]+='There is a door to the West. '

        return tuple(to_return)
````

````python
class Finish(Room):
    def __init__(self, ne, ee, se, we, finish_text):
        super().__init__(ne, ee, se, we)

        self.finish_text = finish_text
    
    def get_finish_text(self):
        return self.finish_text
    def set_finish_text(self, data):
        self.finish_text = data

    def get_message(self):
        return "Finish", self.get_finish_text()
````

````python
class Normal(Room):
    def __init__(self, ne, ee, se, we, qs, ans):
        super().__init__(ne, ee, se, we)

        self.question = qs
        self.answer = ans
    
    def get_question(self):
        return self.question
    def set_question(self, data):
        self.question = data

    def get_answer(self):
        return self.answer
    def set_answer(self, data):
        self.answer = data

    def check_answer(self, ans):
        print(self.answer, ans)
        return self.answer==ans
    
    def get_message(self):
        to_return = ["Normal", '']

        if self.get_exit('N') != -1:
            to_return[1]+='There is a door to the North. '
        
        if self.get_exit('S') != -1:
            to_return[1]+='There is a door to the South. '
        
        if self.get_exit('E') != -1:
            to_return[1]+='There is a door to the East. '

        if self.get_exit('W') != -1:
            to_return[1]+='There is a door to the West. '

        return tuple(to_return)
````

  * room type: Start, Finish or Normal
  * the room number that can be accessed to the North of the room, or -1 if there is no door
  * the room number that can be accessed to the East of the room, or -1 if there is no door
  * the room number that can be accessed to the South of the room, or -1 if there is no door
  * the room number that can be accessed to the West of the room, or -1 if there is no `door`
  * the start message if the room is of type Start
    * or the finish message if the room is of type `Finish`
    * or the question if the room is of type `Normal`
  * the answer to the question if the room is of type `Normal`

The rooms are read in the order they appear in the file and stored in a 1-dimensional list. For example: the room in the first row of the file is stored in index 0; the room in the second row of the file is stored in index 1.

Write program code for a function to:

  * read the data from the text file
  * create an object of the appropriate class for each room
  * store the rooms in a 1-dimensional list
  * return the 1-dimensional list. <div style="text-align: right">[7]</div>

````python
# Task 2.2
def make_rooms():
    import csv

    with open('./resources/rooms.txt') as f:
        data = [[int(j) if j.lstrip('-').isnumeric() else j for j in i] for i in csv.reader(f)]

    rooms = []

    for row in data:
        if row[0] == 'Start':
            rooms.append(Start(row[1], row[2], row[3], row[4], row[5]))
        
        elif row[0] == 'Normal':
            rooms.append(Normal(row[1], row[2], row[3], row[4], row[5], row[6]))
        
        elif row[0] == 'Finish':
            rooms.append(Finish(row[1], row[2], row[3], row[4], row[5]))
    
    return rooms    

make_rooms()
````

Three tables in the database are shown:

```text
User (Username*, Blocked)
Post (PostID*, PostText, DateTimePosted, Username^)
Comment (CommentID*, PostID^, Username^, TheText, DateTimeCommented)
```

A primary key is indicated by the `*` at the end of the attribute name. Foreign keys are indicated by using a `^`.

The database is being created and tested using sample data which is given in the following text files: `user.txt`, `post.txt` and `comment.txt`

The field `Blocked` in the table `User` stores 1 if the user is currently blocked and stores 0 if the user is not blocked.

## Task 3.1

Write a Python program that uses SQL to create a database with the three tables: `User`, `Post` and `Comment`. Define the primary and foreign keys for each table. <div style="text-align: right">[6]</div>

````python
# Task 3.1
import sqlite3
import csv

conn = sqlite3.connect('website.db')
cursor = conn.cursor()

cursor.executescript('''
    CREATE TABLE IF NOT EXISTS User(
        Username TEXT PRIMARY KEY,
        Blocked INT NOT NULL DEFAULT 0                 
    );
                     
    CREATE TABLE IF NOT EXISTS Post(
        PostID TEXT PRIMARY KEY,
        PostText TEXT NOT NULL,
        DateTimePosted TEXT NOT NULL,
        Username TEXT NOT NULL,

        FOREIGN KEY (Username) REFERENCES User(Username)                 
    );
    
    CREATE TABLE IF NOT EXISTS Comment(
        CommentID INTEGER PRIMARY KEY,
        PostID TEXT NOT NULL,
        Username TEXT NOT NULL,
        TheText TEXT NOT NULL,
        DateTimeCommented TEXT NOT NULL,

        FOREIGN KEY (PostID) REFERENCES Post(PostID),
        FOREIGN KEY (Username) REFERENCES User(Username)       
    );
''')

conn.commit()
conn.close()
````

A computer program prototype is being developed to store records in a hash table. 

Each record has a key field (positive integer) and a string data item.

The hash table is set up as a list, initialised with capacity for 100 records.

The hashing algorithm is: `hash(k) = k mod s`

`k` represents the key field and `s` represents the maximum size of the hash table.

## Task 4.1
Write program code to initialise the global list `hash_table` with 100 empty records. <div style="text-align: right">[1]</div>

````python
# Task 4.1
hash_table = [None for i in range(100)]
````

## Task 4.2

Write program code for the function `calculate_hash()` to take a key field as a parameter and return the calculated hash value using the given hashing algorithm. <div style="text-align: right">[2]</div>

````python
# Task 4.2
def calculate_hash(k):
    return k % len(hash_table)
````

## Task 4.3
The hash table uses linear probing when there is a collision. Linear probing is when the next index is checked to identify whether the index is free. This repeats until either a free index is found, or all indices have been checked and the list is full.

Write program code for the function `insert()` to:

  * take a key field and string data item as parameters
  * calculate the index using the `calculate_hash()` function
  * store the key field and string data item in the index if available, otherwise use linear probing to find the next free index
  * return `True` if the record was successfully stored or `False` if the hash table is full. <div style="text-align: right">[6]</div>

````python
# Task 4.3
def insert(k, data):
    index = calculate_hash(k)
    for i in range(len(hash_table)):
        index = (index+i) % len(hash_table)

        if not hash_table[index]:
            hash_table[index] = (k ,data)
            return True
        
    return False
````

````python
# Task 4.4
import csv

def read_data():
    with open('./resources/records.txt') as f:
        data = [i for i in csv.reader(f, delimiter=' ')]
        for rec in data:
            success = insert(int(rec[0]), rec[1])
            if not success:
                print(f'Record: {rec[1]}, Cannot be inserted')
````

## Task 4.5

Write program code for the function `find_record()` to:

  * take a key field as a parameter

````python
# Task 4.5
def find_record(k):
    index = calculate_hash(k)
    for i in range(len(hash_table)):
        index = (index+i) % len(hash_table)

        if not hash_table[index]:
            break

        if hash_table[index][0] == k:
            return hash_table[index][1]
        
    return False
````
