The IMEI number is a unique value used to identify mobile devices, such as phones and tablets.
The IMEI number has 15 digits: 14 digits with an extra check digit added to the right-hand side.

The check digit is calculated using the following algorithm, on the left-most 14 digits of an IMEI number:
1. starting from the right, the first digit is location number 1
2. double all digits in the odd numbered positions
3. sum all the digits, including both the unchanged digits (i.e. those in the even numbered positions) as well as those doubled (e.g. 16 contributes 1 + 6)
4. the check digit is the value between O and 9 that must be added to the sum to make the esult exactly divisible by 10.

For example, given the 14 digits 14576567654934:

<center>

||||||||||||||||
|---|---|---|---|---|---|---|---|---|---|---|---|---|---|---|
| Step 1: | 1 | 4 | 5 | 7 | 6 | 5 | 6 | 7 | 6 | 5 | 4 | 9 | 3 | 4 |
| Step 2: | 1 | 8 | 5 | 14 | 6 | 10 | 6 | 14 | 6 | 10 | 4 | 18 | 3 | 8 |
| Step 3: | 1+ | (8)+ | 5+ | (1+4)+ | 6+ | (1+0)+ | 6+ | (1+4)+ | 6+ | (1+0)+ | 4 | (1+8)+ | 3 | (8) |

</center>

Sum= 68

Step 4: Check digit = 2


## Task 1.1
Write a function `task1_1(input_value)` that returns an integer.
The function should:
- validate that the parameter `input_value` is either an integer, or a string containing a valid integer
- check that it is 14 digits in length
- return -1 if the value received is invalid for any reason
- calculate the check digit for the given first 14 digits of the IMEI number
- return the calculated check digit. 

<div style="text-align: right">[6]</div> 

````python
# YOUR ANSWER HERE
def task1_1(input_value):
    if not type(input_value) == int:
        try:
            input_value = int(input_value)
        except:
            return -1
        
    if len(str(input_value)) != 14: return -1

    sum_list = []
    for i in range(14):
        n = int(str(input_value)[i])
        if i % 2 != 0: # since 0 indexing
            n = n*2
        sum_list.append(n)

    summation = 0
    for n in sum_list:
        if len(str(n)) > 1:
            summation+=int(str(n)[0]) + int(str(n)[1])
        else:
            summation+=n

    return 10 - summation % 10

task1_1(14576567654934)
````

## Task 2.2
One method of sorting is the insertion sort.
Write a function `task2_2(list_ of_integers)` that:
- takes a list of integers
- implements an insertion sort algorithm
- returns the sorted list of integers.


Call your function `task2_2` with the contents of the file `TEN.txt`, printing the returned list, for example, using the following statement:
```
print(task2 2(task2 l('TEN.txt'))) 
```
<div style="text-align: right">[1]</div> 

````python
# YOUR ANSWER HERE
def task2_2(list_of_integers):
    for i in range(1, len(list_of_integers)):
        key = list_of_integers[i]
        j = i-1

        while j>=0 and list_of_integers[j] > key:
            list_of_integers[j+1] = list_of_integers[j]
            j-=1
        
        list_of_integers[j+1] = key
    
    return list_of_integers

print(task2_2(task2_1('TEN.txt'))) 
````

## Task 2.3
Another method of sorting is the quicksort. 

Write a function `task2_3(list_of_integers)` that:

Call your function `task2_3` with the contents of the file `TEN.txt`, printing the returned list, for example, using the following statement:

<div style="text-align: right">[1]</div> 

````python
# YOUR ANSWER HERE
def task2_3(list_of_integers):
    if len(list_of_integers) <=1:
        return list_of_integers
    
    pivot = list_of_integers[len(list_of_integers) // 2]

    less = [i for i in list_of_integers if i < pivot]
    equal = [i for i in list_of_integers if i == pivot]
    more = [i for i in list_of_integers if i > pivot]

    return task2_3(less) + equal + task2_3(more)

print(task2_3(task2_1('TEN.txt') ))
````

A programmer is writing a class, `LinkedList`, to represent a linked list of unique integers. A linked list is a collection of data elements, whose order is not given by their physical placement in memory. Instead, each element points to the next.

## Task 3.1
Write the `LinkedList` class in Python. Use of a simple Python list is not sufficient. Include the following methods:
- `insert(integer_value)` inserts the `integer_value` at the beginning (head) of the list
- `delete(integer_value)` attempts to delete `integer_value` from the list; if the item was not present, return `None`
- `search(integer_value)` returns a Boolean value: `True` if `integer_value` is in the list, `False` if not in the list
- `count()` should return the number of elements in the list, or zero if empty 
- `to_String()` should return a string containing a suitably formatted list with the elements separated by a comma and a space, with square brackets at either end, e.g. in the form:
    ```
    [11, 2, 7, 4]
    ```
<div style="text-align: right">[8]</div> 

Test `LinkedList` by using the data in the file `Task3data. txt`. Use the `to_String()` method to print the resulting contents of the list. 
<div style="text-align: right">[3]</div> 

````python
# YOUR ANSWER HERE
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

    def get_data(self):
        return self.data
    
    def set_data(self, data):
        self.data = data

    def get_next(self):
        return self.next
    
    def set_next(self, next):
        self.next = next

    def __repr__(self):
        return self.data

class LinkedList:
    def __init__(self):
        self.first = None

    def get_first(self):
        return self.first
    
    def set_first(self, first):
        n = Node(first)
        self.first = n

    def insert(self, integer_value):
        if self.search(integer_value): return

        second = self.first
        self.set_first(integer_value)
        self.first.set_next(second)

    def delete(self, integer_value):
        if not self.first: return

        if self.first.get_data() == integer_value:
            self.first = self.first.get_next()

        current = self.first
        previous = None

        while current and current.get_data() != integer_value:
            previous = current
            current = current.get_next()

        if not current:
            return
        
        previous.next = current.next

    def search(self, integer_value):
        if not self.first: return False

        current = self.first

        while current and current.get_data() != integer_value:
            current = current.get_next()
            if current and current.get_data() == integer_value:
                return True
            
        return False
    
    def count(self):
        if not self.get_first(): return 0
        current = self.first

        c = 0
        while current:
            current = current.get_next()
            c+=1
        
        return c
    
    def to_String(self):
        if not self.get_first(): print('[]'); return
        current = self.first

        c = []
        while current:
            c.append(current.data)
            current = current.get_next()

        return str(c)

with open('./resources/Task3data.txt') as f:
    LL = LinkedList()

    for i in f.readlines():
        LL.insert(int(i.strip()))

LL.to_String()
````

Additional `enqueue` and `dequeue` methods are to be defined on the `Queue` class:

````python
# YOUR ANSWER HERE
class Queue(LinkedList):
    def __init__(self):
        super().__init__()

    def enqueue(self, integer_value):
        if not self.first: self.set_first(integer_value); return

        current = self.first
        previous = None

        while current:
            previous = current # keep the previous because current will be Nonetype
            current = current.get_next()
        
        n = Node(integer_value)
        previous.next = n
        return
    
    def dequeue(self):
        r = self.first # store the first to return later
        self.first = self.first.next

        return r.get_data()
    
with open('./resources/Task3data.txt') as f:
    LL = Queue()

    for i in f.readlines():
        LL.enqueue(int(i.strip()))

LL.to_String()
for i in range(5):
    dq = LL.dequeue()
    print(dq)
````
